# 錯誤更正碼 0914

## 通訊原理

### 通訊用在哪裡？


通訊的目的  
是把某個地方的資料送到另一個地方  

- 藍牙：滑鼠、鍵盤、耳機  
- Wi-Fi 與行動網路：手機、電腦、基地台  
- 光纖與海底電纜：跨城市、跨國的資料傳輸  
- 衛星與深空通訊：距離很遠，也不容易重傳  

不論距離長短  
接收端都要從不完美的物理訊號中還原資料


### 儲存也是一種通訊


儲存可以想成  
把資料從「現在」傳到「未來」  

```text
資料 → 寫入 → 儲存媒介 → 讀取 → 資料
資料 → 發送 → 通訊媒介 → 接收 → 資料
```

硬碟、記憶體、光碟與 DNA  
都需要把抽象的符號變成實際的物理狀態  
再把物理狀態讀回符號  
這兩個方向都可能出錯


### 資料的媒介：磁性


硬碟利用磁性材料的磁化狀態儲存資料  
讀取頭感測這些磁化狀態產生的磁場變化  

**巨磁阻（giant magnetoresistance, GMR）**  
是磁性層的相對磁化方向改變時  
電阻會**明顯**改變的現象  
因此可以把磁性資訊轉成電訊號  

磁化區域做得越小  
同一片碟片就能放越多資料  
但讀取時也需要分辨更微弱的訊號  

參考：[GMR 與硬碟讀取（Nobel lecture）](https://www.nobelprize.org/uploads/2018/06/grunberg-slides.pdf)


### 資料的媒介：電荷


「用電量儲存」可以更精確地說成  
利用儲存的**電荷**改變可量測的電壓或電晶體狀態  

- **DRAM**  
    利用電容上的電荷儲存資料  
    電荷會流失，因此需要定期更新  
- **Flash memory**  
    利用浮動閘極或電荷捕捉層中的電荷  
    改變電晶體的臨界電壓  
    SSD 與隨身碟都使用這類技術  

可以把不同的電壓區間對應到不同資料  
區間分得越細  
一次能存的 bit 越多  
但受到雜訊干擾時也越容易判錯


### 資料的媒介：凹坑

以壓製的光碟為例  
資料記錄在塑膠基板上的 **pit（凹坑）** 與 **land（平面）** 結構  
再配合反射層與雷射讀取  

讀取端觀察反射光的變化  
經過訊號處理與解碼後還原資料  
並不是直接把「每個坑」當成一個 1  

刮傷、灰塵或讀取不穩定  
可能讓一整段資料受到影響  
這也是光碟需要錯誤更正碼的原因


### 資料的媒介：DNA


DNA 使用四種鹼基  
$$A, \ T, \ C, \ G.$$

四種符號在不加限制時  
每個位置可以表示 $\log_2 4 = 2$ 個 bits  
實際設計還要考慮合成、讀取與序列限制  

```text
bits → 鹼基序列 → 合成 DNA → 保存 → 定序 → bits
```

錯誤可能包含替換、插入、刪除  
也可能整條 DNA 都沒有被讀到


#### Illumina 怎麼讀 DNA？


Illumina 的 **sequencing by synthesis（邊合成邊定序）**  
在合成互補鏈的過程中  
利用帶有螢光標記的可逆終止核苷酸  
逐輪讀取加入的鹼基  

以光激發螢光標記  
拍攝訊號，再推回鹼基序列  
這裡是**螢光讀取**   

參考：[Illumina sequencing technology](https://www.illumina.com/science/technology/next-generation-sequencing/sequencing-technology.html)


### 資料的媒介：光子（光纖／光纜）


光纖通訊利用光攜帶資料  
從粒子的觀點看  
光的能量由光子傳遞  

發送端把資料編成光訊號  
讓光沿著光纖傳播  
接收端再用光偵測器轉成電訊號  

光纜把光纖包覆、保護起來  
海底光纜就是長距離通訊的例子  
傳統金屬電纜則利用電訊號傳輸  

傳輸過程中的衰減與接收端雜訊  
都可能讓資料判讀出錯


### 資料的媒介：電磁波（無線通訊）


無線通訊利用天線發送與接收電磁波  
讓資料經由空間傳播  
例如藍牙、Wi-Fi、行動網路與衛星通訊  

可以改變波的振幅、頻率或相位  
讓不同的訊號代表不同資料  
接下來的 BPSK 與 QPSK 就是利用相位傳遞資訊  

光子就是電磁波就是光子  
但是光纖內的環境相對穩定，可以精細到個別光子
無線通訊就稍微混亂

無線訊號可能受到雜訊、干擾與多重路徑影響  
接收端需要從這些變化中還原資料

## 無線通訊


### 從 bits 到電磁波


天線發送的是連續變化的電磁波  
因此需要先把 bits 對應到不同的波型  
這個步驟稱為 **modulation（調變）**  

```text
資訊 bits → 錯誤更正編碼 → 調變 → 通道
                                  ↓
還原資訊 ← 錯誤更正解碼 ← 解調 ← 接收訊號
```

先考慮載波相位可以同步  
且通道增益已經補償的簡化情況


### BPSK：兩種相位


**Binary Phase-Shift Keying**  
用相差 $\pi$ 的兩種相位代表一個 bit  
例如  
$$0 \longmapsto +A, \qquad 1 \longmapsto -A.$$

對應的載波可以寫成  
$$s_b(t) = A\cos(2\pi f_ct + \pi b), \qquad b \in \{0, 1\}.$$

星座圖上只有兩個點  

```text
         1                   0
─────────●─────────┼─────────●───────── I
        −A         0        +A
```

接收端估計訊號落在哪一側  
每個 symbol 攜帶 1 個 bit


### QPSK：四種相位


**Quadrature Phase-Shift Keying**  
使用四種相位  
每個 symbol 攜帶 2 個 bits  

在複數基頻平面上  
一種 Gray labeling 是  
$$
\begin{aligned}
00 & \longmapsto (A, A), & 01 & \longmapsto (-A, A), \\
11 & \longmapsto (-A, -A), & 10 & \longmapsto (A, -A).
\end{aligned}
$$

```text
                     Q
                     ↑
        01 ●         │         ● 00
                     │
        ─────────────┼─────────────→ I
                     │
        11 ●         │         ● 10
```

$I$ 與 $Q$ 是兩個互相正交的分量  
相鄰星座點的標籤只差 1 個 bit  
因此誤判到鄰點時只會錯 1 個 bit


### 每個 symbol 多放一個 bit（？）


在相同 symbol rate 下  
QPSK 的原始 bit rate 是 BPSK 的兩倍  
但也不是說一直往星座圖裡加點點就是好事

比較抗雜訊能力時  
必須固定能量的比較基準  
以上面為例，BPSK 點能量是 $A^2$  
QPSK 點能量是 $2A^2$  

### 為什麼會出現 Gaussian noise？


接收端會在一段時間內累積訊號  
例如與載波相乘後積分  
可以理解成取加權平均  

若雜訊由許多微小、近似獨立的因素相加而成  
且滿足有限變異數等中央極限定理需要的條件  
標準化後的總和可以近似成高斯分布  

例如 $N_i$ 獨立同分布  
平均為 0，變異數為 $\sigma^2$  
則  
$$\frac{N_1 + \cdots + N_m}{\sigma \sqrt m}
\xrightarrow{d} \mathcal N(0, 1).$$

平均雜訊的變異數則是  
$$\operatorname{Var} \left(\frac1m \sum_{i = 1}^m N_i \right)
= \frac{\sigma^2}{m}.$$


### AWGN 的三個假設


**Additive White Gaussian Noise**  

- **Additive：加性**  
    雜訊直接加在訊號上  
    $Y = X + N$  
- **White：白雜訊**  
    在考慮的頻帶內，功率頻譜密度是平坦的  
- **Gaussian：高斯**  
    雜訊以高斯隨機過程建模  

剛剛說的「取平均」機制能在適當條件下解釋高斯  
但不能單靠取平均就推出加性與白雜訊  
濾波後的雜訊也可能具有時間相關性  

參考：[MIT 6.02：Noise](https://ocw.mit.edu/courses/6-02-introduction-to-eecs-ii-digital-communication-systems-fall-2012/56d958fded205681d34fab3eac4b7ab5_MIT6_02F12_chap09.pdf)


### 從 AWGN 到 bit flip


BPSK 的一個實數接收樣本可以寫成  
$$Y = X + N, \qquad X \in \{-A, +A\}, \quad N \sim \mathcal N(0, \sigma^2).$$

在兩個 bits 等機率的情況下  
以 0 為門檻做 hard decision  
$$
\widehat b = \begin{cases}
0, & Y \ge 0, \\
1, & Y < 0.
\end{cases}
$$

雜訊把點推過門檻時  
就會發生 bit flip  
錯誤機率是  
$$p = Q\!\left(\frac A\sigma \right), \qquad
Q(x) = \frac1{\sqrt{2\pi}}\int_x^\infty e^{-t^2/2}\,dt.$$

若各次 decision 的雜訊獨立且同分布  
就得到之前使用的 binary symmetric channel（BSC）模型  
也可以保留接收值的大小作為可靠度資訊（AWGNC）  
交給 soft-decision decoder

## 更多漢明碼


### 從 $[7, 4, 3]$ 推廣


上次使用 3 個 parity checks  
可以產生 $2^3 = 8$ 種 syndromes  

其中一種代表沒有錯  
其餘 7 種分別指出 7 個可能的單一錯誤位置  

如果改用 $r$ 個 parity checks 呢？  
很自然的我們會想到
$$ n = 2^r - 1. $$

以下討論 binary Hamming code  
並假設 $r \ge 2$


### Parity-check matrix


延續上次的 row-vector 慣例  
令 $H$ 是 $n \times r$ 矩陣  
第 $j$ 列是 $j$ 的二進位表示  
也就是把 $\mathbb F_2^r$ 的所有非零向量  
按照字典序各排一次  

接著定義
$$C = \{v \in \mathbb F_2^n : vH = 0\}. $$

所有運算都在 $\mathbb F_2$ 中進行  
加法就是 XOR  

$H$ 包含 $r$ 個標準基底向量  
所以 $\operatorname{rank}(H) = r$  
由 rank-nullity theorem  
$$ k = n - r = 2^r - 1 - r.$$


### 一般的 systematic encoding


把 parity bits 放在編號為 2 的冪次的位置  
$$1, 2, 4, \ldots, 2^{r - 1}.$$

其他 $n - r$ 個位置放資訊 bits  
因為這些 parity 位置的 $H$ 組成一個單位矩陣  
可以唯一決定 parity bits  
使得 $vH = 0$  

例如 $r = 3$ 時  
就是上一節課的排法  
$$ (p, q, a, r', b, c, d), $$
其中 $r'$ 是第 4 個位置的 parity bit  
避免和 parity checks 的數量 $r$ 混淆


@@@ what about r = 4 @@@

### Syndrome decoding


假設傳送 $v \in C$  
收到 $u = v + e$  
則  
$$uH = (v + e)H = eH.$$

如果只有第 $j$ 個 bit 出錯  
$eH$ 就等於 $H$ 的第 $j$ 列  
也就是位置 $j$ 的二進位表示  

- syndrome 為 0：不更動  
- syndrome 非零：把所指的位置翻轉  

因此可以更正任意一個 bit flip  
多個錯誤的 syndrome 可能看起來像單一錯誤  
這時就可能改錯位置


### Hamming code 的 minimum distance


對 linear code  
兩個 codewords 的差仍是 codeword  
所以 minimum distance 等於最小非零 weight  
$$d = \min_{v \in C, \ v \ne 0}\operatorname{wt}(v).$$

weight 就是向量中 1 的數量  
要找 weight 很小的 codeword  
等同於找少數幾列 $H$，使它們相加為 0


### 為什麼 $d = 3$？


$H$ 沒有零列  
所以不存在 weight 1 的 codeword  

$H$ 沒有重複的列  
所以不存在 weight 2 的 codeword  
因此 $d \ge 3$  

另一方面  
編號 1、2、3 的三列滿足  
$$(00 \cdots 001) + (00 \cdots 010) + (00 \cdots 011) = 0.$$
只在位置 1、2、3 放 1 的向量就是 codeword  
所以 $d \le 3$  

因此  
$$\boxed{C\text{ 的參數為 }[2^r - 1, \ 2^r - 1 - r, \ 3].}$$


### 所有的 Hamming code 都是 perfect code


每個半徑 1 的 Hamming sphere 有  
$$1 + n = 2^r$$
個向量  
總共有 $2^k = 2^{n - r}$ 個 codewords  
因此  
$$2^k(1 + n) = 2^{n - r}2^r = 2^n.$$
$d = 3$ 保證這些球不重疊  
而等號表示它們剛好覆蓋所有 binary vectors  

| $r$ | $n$ | $k$ | $d$ |
|---|---|---|---|
| 2 | 3 | 1 | 3 |
| 3 | 7 | 4 | 3 |
| 4 | 15 | 11 | 3 |
| 5 | 31 | 26 | 3 |


## Extended Hamming code


### 再加一個 overall parity bit


對每個 Hamming codeword $v = (v_1, \ldots, v_n)$  
加上  
$$p(v) = v_1 + \cdots + v_n \pmod 2.$$
得到  
$$\widetilde v = (v_1, \ldots, v_n, p(v)).$$
這樣每個新 codeword 都有偶數個 1  
新加入的 bit 完全由原本的 codeword 決定  
因此長度增加 1  
資訊量 $k$ 不變  
$$\widetilde n = 2^r, \qquad \widetilde k = 2^r - 1 - r.$$


### Extended parity-check matrix


使用相同的 row-vector 慣例  
可以把新的矩陣寫成  
$$
\widetilde H =
\begin{bmatrix}
H & \mathbf 1_n\\
\mathbf 0_{1 \times r} & 1
\end{bmatrix}.
$$
前 $r$ 個 checks 檢查原本的 Hamming code  
最後一個 check 檢查全部 $n + 1$ 個 bits 的 parity  
$$\widetilde C = \{\widetilde v : \widetilde v\widetilde H = 0\}.$$

新的一欄與原本的 $r$ 欄獨立  
所以 rank 是 $r + 1$  
也得到 dimension $2^r - r - 1$


@@@ r = 3 example of H tilde e@@@

### Distance of extended Hamming code


令原本 codeword 的 weight 是 $w$  
加上 overall parity bit 後  
$$\operatorname{wt}(\widetilde v) =
\begin{cases}
w, & w\text{ 是偶數}, \\
w + 1, & w\text{ 是奇數}.
\end{cases}$$

原本非零 codeword 的 weight 至少是 3  
weight 3 會變成 4  
weight 至少 4 的也不會變小  
所以 $\widetilde d \ge 4$  


而原本確實存在 weight 3 的 codeword  
延伸後正好是 weight 4  
所以 $\widetilde d \le 4$  
$$\boxed{\widetilde C\text{ 的參數為 }
[n, k, d] = [2^r, 2^r - 1 - r, 4].}$$


#### $[7, 4, 3]$ 變成 $[8, 4, 4]$


原本的 codeword  
$$v = (1, 1, 1, 0, 0, 0, 0)$$
滿足 $vH = 0$  
weight 是 3  
加上 overall parity bit 後  
$$\widetilde v = (1, 1, 1, 0, 0, 0, 0, 1)$$
weight 變成 4  
因此它和全零 codeword 的距離正好是 4  
這也具體展示了 minimum distance 的上界


### 多一個 parity bit 有什麼用？


$d = 4$ 仍然只能保證更正 1 個錯誤  
$$\left\lfloor\frac{4 - 1}{2} \right\rfloor = 1.$$

但可以做到 **SECDED**  
也就是 single-error correction, double-error detection  

把收到的向量寫成 $(u, q)$  
計算  
$$s = uH, \qquad b = \sum_{i = 1}^n u_i + q \pmod 2.$$

**假設總共至多 2 個 bit flips**  

| Hamming syndrome $s$ | Overall parity $b$ | 判斷與處理 |
|---|---|---|
| $0$ | $0$ | 沒有錯誤 |
| 非零 | $1$ | 原本的 $n$ 個 bits 中有一個錯誤，翻轉 $s$ 所指的位置 |
| $0$ | $1$ | 只有新增的 parity bit 出錯，翻轉它 |
| 非零 | $0$ | 兩個 bits 出錯，回報偵測到雙錯誤 |


#### 偵測與更正的保證要分開看


若只檢查收到的向量是不是 codeword  
$d = 4$ 可以偵測任意 1、2、3 個 bit flips  

但執行上面的 SECDED 解碼時  
3 個 bit flips 可能被誤認為單一錯誤  
因此不能宣稱同時「更正 1 個並偵測所有 3 個」  

| Code | 參數 | 保證更正 | 只做合法性檢查時保證偵測 |
|---|---|---|---|
| Hamming | $[2^r - 1, 2^r - 1 - r, 3]$ | 1 個錯誤 | 至多 2 個錯誤 |
| Extended Hamming | $[2^r, 2^r - 1 - r, 4]$ | 1 個錯誤 | 至多 3 個錯誤 |
