### $\mathrm{RM}(2, 5)$ 的 randomized erasure decoder

考慮 binary Reed–Muller code  
$$\mathcal C = \mathrm{RM}(2, 5) \subseteq \mathbb F_2^{32}.$$
它是 self-dual code：  
$$\mathcal C^\bot = \mathrm{RM}(2, 5) = \mathcal C.$$
因此每個 weight 為 $8$ 的 codeword 也能當作 parity check

先隨便固定一個 minimum-weight parity check $h_0 \in \mathcal{C}^\bot$  
例如 $x_1x_2$  
下列程序可以隨機產生一個 minimum-weight parity check：

1. 均勻隨機產生一個 $5 \times 5$ binary matrix $A$
2. 若 $A$ 不可逆，就重生，直到 $A \in \operatorname{GL}(5, 2)$
3. 均勻隨機產生 $b \in \mathbb F_2^5$
4. 用 affine permutation $x \mapsto Ax + b$ 重新排列 $h_0$ 的 coordinates

考慮通道 $\operatorname{BEC}(p)$  
對收到的 received word 重複以下步驟 $1{,}000$ 次：

1. 以上述隨機法生一個 weight-$8$ parity check
2. 若 PC 的 $8$ 個 coordinates 中恰有一個 erasure，  
   就利用 parity-check equation 補回該 coordinate
3. 否則不做任何事，繼續下一次嘗試

若 $1{,}000$ 次後仍有任何 erasure，就宣告 block error

對每個  
$$p \in \{0, 0.1, 0.2, \ldots, 1.0\}$$

以 $p$ 為橫軸、bit/block error probability 為縱軸作圖

### $\mathrm{RM}(2, 5)$ 的 randomized soft decoder

考慮同一個 code $\mathcal C = \mathrm{RM}(2, 5)$，但通道改為 $\operatorname{BSC}(p)$  
利用 linearity 與 BSC 的 symmetry，可以固定傳送 all-zero codeword  
用 spin $s_i = (-1)^{x_i} \in \{+1, -1\}$ 表示每個 transmitted symbol  
將 received symbol 的資訊寫成 bias  
$$\beta_i = \Pr(s_i = +1 \mid y_i) - \Pr(s_i = -1 \mid y_i) \in [-1, 1].$$
因此 BSC output 對應的初始 bias 為  
$$\beta_i = (1 - 2p)(-1)^{y_i}.$$

每次用週五作業的方法隨機產生一個 minimum-weight parity check  
令它的 support 為 $S$，所以 $|S| = 8$  
對每個 $i \in S$，用另外 7 個 coordinates 的 biases 計算  
$$\gamma_i = \prod_{j \in S \setminus \{i\}} \beta_j,$$
再更新  
$$\beta_i \longleftarrow \beta_i + (1 - |\beta_i|)\gamma_i.$$
同一個 check 中的 8 個 $\gamma_i$ 都必須用更新前的 biases 計算，然後同步更新

重複抽取 check 並更新 $1{,}000$ 次，再取每個 $\beta_i$ 的 sign 作為 hard decision  
$\beta_i > 0$ 時判為 $\hat x_i = 0$，$\beta_i < 0$ 時判為 $\hat x_i = 1$  
若 $\beta_i = 0$，就在 $0$ 與 $1$ 之間均勻隨機決定  
將整個 decoded word 與 transmitted codeword 比較  
只要有任一 coordinate 錯誤，就記為一次 block error

對每個  
$$p \in \{0, 0.05, 0.10, \ldots, 0.50\}$$
分別進行 $10{,}000$ 次獨立的 channel realizations，估計 block error probability  
以 $p$ 為橫軸、block error probability 為縱軸作圖  
圖上必須標示座標軸與 random seed  
繳交程式、各個 $p$ 的數值結果與圖片
