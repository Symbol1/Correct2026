# 錯誤更正碼 1102

期中考前的最後一份講義，用 RS codes 得到 MDS 的 $[n, k, n - k + 1]_q$ codes。  
代價是 $n \le q$：取值點必須互不相同。  
但是硬碟、記憶體、通訊最終處理的都是 bits，而 $\mathbb F_2$ 上的 RS code 長度至多是 $2$。  
今天的問題是：能不能保留 RS 的代數結構與 decoder，卻得到長的 **binary** codes？

食譜
- 每個 cyclic code 都由 $z^n - 1$ 的一個 monic 因式決定，稱為 generator polynomial。
- 用 Frobenius map 把 $z^n - 1$ 在 $\mathbb F_2$ 上分解：cyclotomic cosets 與 minimal polynomials。
- BCH bound：連續 $\delta - 1$ 個零點保證 $d \ge \delta$。
- Binary BCH code 就是 RS code 中所有 binary codewords 形成的 subcode。
- RS 的 syndrome decoding 直接適用，而且在 binary 的情形更簡單。

## Cyclic Codes 與 Generator Polynomials

### 回顧：cyclic code 與多項式

1005 講義把長度 $n$ 的向量寫成多項式
$$c = (c_0, \ldots, c_{n-1})
\quad\longleftrightarrow\quad
c(z) = c_0 + c_1 z + \cdots + c_{n-1} z^{n-1},$$
並看成 $\mathbb F_q[z]/(z^n - 1)$ 中的元素。  
Cyclic shift 就是乘上 $z$ 再取模 $z^n - 1$，  
所以 cyclic code 對「乘上任意多項式再取模 $z^n - 1$」封閉。

1012 講義的 cyclic RS code 有 generator polynomial $g(z) = (z - \alpha) \cdots (z - \alpha^{n-k})$，  
codewords 恰好是被 $g(z)$ 整除的多項式。  
這其實是所有 cyclic codes 共同的結構。

### 每個 cyclic code 都有 generator polynomial

令 $\mathcal C$ 是長度 $n$ 的 cyclic code，至少有一個非零 codeword；  
把 codewords 看成次數小於 $n$ 的多項式。  
在 $\mathcal C$ 的非零多項式中，取次數最小的一個，再乘上常數使它 monic，記為 $g(z)$。  
它是唯一的：若有兩個，相減會得到次數更小的非零 codeword。  
沿用 1012 講義的記號 $\mathbb F_q[z]_{<k}$，表示零多項式與次數小於 $k$ 的多項式。

**定理：** 令 $k = n - \deg g$，則
1. $\mathcal C = \{m(z) g(z) : m \in \mathbb F_q[z]_{<k}\}$，也就是 codewords 恰好是 $g(z)$ 的倍式。
2. $g(z)$ 整除 $z^n - 1$。
3. $\dim \mathcal C = k$。

證明：任取 codeword $c(z)$，做多項式除法 $c(z) = m(z) g(z) + r(z)$，  
其中 $r = 0$ 或 $\deg r < \deg g$。  
$m(z) g(z) = c(z) - r(z)$ 的次數小於 $n$，不需要取模，由封閉性它屬於 $\mathcal C$；  
因此 $r = c - mg$ 也屬於 $\mathcal C$。  
由 $g$ 的最小性，$r = 0$。  
反過來，$m \in \mathbb F_q[z]_{<k}$ 時 $m(z) g(z)$ 的次數小於 $n$，同樣由封閉性屬於 $\mathcal C$。

把 $z^n - 1$ 除以 $g(z)$，寫成 $z^n - 1 = h(z) g(z) + r(z)$。  
在模 $z^n - 1$ 之下，$r(z) \equiv -h(z) g(z)$，由封閉性屬於 $\mathcal C$，再由最小性得到 $r = 0$。  
最後，$m \mapsto mg$ 是一一的線性映射，$m$ 有 $k$ 個係數，所以 $\dim \mathcal C = k$。

反過來，任取整除 $z^n - 1$ 的 monic polynomial $g(z)$，令 $k = n - \deg g$，  
$\{m(z) g(z) : m \in \mathbb F_q[z]_{<k}\}$ 都是 cyclic code：  
乘上 $z$ 後若次數達到 $n$，就減去 $z^n - 1$ 的常數倍，結果仍是 $g$ 的倍式。

因此長度 $n$ 的 cyclic codes 與 $z^n - 1$ 的 monic 因式一一對應：
$$\mathcal C \longleftrightarrow g(z),\qquad \dim \mathcal C = n - \deg g.$$
約定 $g = z^n - 1$ 對應只有零向量的 code。  
要列出所有 cyclic codes，就要先分解 $z^n - 1$。

### 用零點描述 cyclic code

接下來固定 $n = 2^m - 1$，令 $\alpha$ 是 $\mathbb F_{2^m}$ 的 primitive element。  
1005 講義證明了 $x^q - x = \prod_{a\in\mathbb F_q}(x - a)$；  
取 $q = 2^m$，除以 $a = 0$ 對應的因式 $x$，得到
$$z^n - 1 = \prod_{j=0}^{n-1}(z - \alpha^j).$$
它的 $n$ 個根 $1, \alpha, \ldots, \alpha^{n-1}$ 互不相同。  
在 $\mathbb F_2$ 中 $-1 = 1$，所以 $z^n - 1$ 也可以寫成 $z^n + 1$。

在 $\mathbb F_{2^m}$ 上看，$z^n - 1$ 的 monic 因式就是一部分 $(z - \alpha^j)$ 的乘積。  
令 $T \subseteq \{0, 1, \ldots, n - 1\}$ 是 $g$ 的根的位置，  
$g(z) = \prod_{j\in T}(z - \alpha^j)$。  
因為這些根互不相同，$g$ 整除 $c$ 當且僅當 $c$ 在每個 $\alpha^j$（$j \in T$）都為零：
$$c \in \mathcal C
\quad\Longleftrightarrow\quad
c(\alpha^j) = 0\quad\text{for all }j \in T.$$
$\{\alpha^j : j \in T\}$ 稱為 $\mathcal C$ 的 **zeros（零點）**，$\dim \mathcal C = n - |T|$。  
為了簡潔，也直接用指數的集合 $T$ 表示 zeros。

反過來說，有 $T$ 就會有 $g$ 嗎？  
別忘了我們要的是 binary code：  
codewords 屬於 $\mathbb F_2^n$，generator polynomial 也必須屬於 $\mathbb F_2[z]$。  
任意挑選 $T$，得到的 $g(z)$ 通常有不在 $\mathbb F_2$ 中的係數，  
例如 $|T| = 1$ 會跑出 $ g(z) = z - \alpha$，常數項係數直接爆掉。  
下一節回答：哪些 $T$ 會給出 $\mathbb F_2[z]$ 中的 $g(z)$？

## Cyclotomic Cosets 與 Minimal Polynomials

### 係數在 $\mathbb F_2$ 中的條件

1005 講義的 Frobenius map $\varphi(a) = a^2$ 是 $\mathbb F_{2^m}$ 的 automorphism。  
被 $\varphi$ 固定的元素滿足 $a^2 = a$，也就是 $x^2 - x$ 的根，只有 $0$ 與 $1$。  
所以
$$a \in \mathbb F_2\quad\Longleftrightarrow\quad a^2 = a.$$

對 $g(z) = \sum_i g_i z^i \in \mathbb F_{2^m}[z]$，把 $\varphi$ 作用在每個係數上，記為 $g^\varphi(z) = \sum_i g_i^2 z^i$。  
$\varphi$ 保留加法與乘法，所以
$$g(z) = \prod_{j\in T}(z - \alpha^j)
\quad\Longrightarrow\quad
g^\varphi(z) = \prod_{j\in T}(z - \alpha^{2j}).$$
因此 $g \in \mathbb F_2[z]$ 當且僅當 $g^\varphi = g$，  
也就是 $\{\alpha^j : j \in T\} = \{\alpha^{2j} : j \in T\}$，  
也就是 $T$ 在 $j \mapsto 2j \bmod n$ 之下封閉。

### Cyclotomic cosets

對 $j \in \{0, 1, \ldots, n - 1\}$，定義 **cyclotomic coset**
$$C_j = \{j, 2j, 4j, \ldots, 2^{m-1} j\} \bmod n.$$
因為 $2^m j = j + nj \equiv j$，再乘一次 $2$ 就回到 $j$。  
Cyclotomic cosets 把 $\{0, 1, \ldots, n - 1\}$ 分割成互不相交的子集合，  
而 $T$ 在乘以 $2$ 之下封閉，當且僅當 $T$ 是一些 cyclotomic cosets 的聯集。

$n = 7$（$m = 3$）時：
$$C_0 = \{0\},\qquad C_1 = \{1, 2, 4\},\qquad C_3 = \{3, 6, 5\}.$$
$n = 15$（$m = 4$）時：
$$C_0 = \{0\},\quad C_1 = \{1, 2, 4, 8\},\quad C_3 = \{3, 6, 12, 9\},\quad C_5 = \{5, 10\},\quad C_7 = \{7, 14, 13, 11\}.$$
$C_5$ 只有兩個元素，因為 $4 \cdot 5 = 20 \equiv 5$。  
同一個 coset 中的元素有相同的 coset，例如 $C_6 = C_{12} = C_3$。

### Minimal polynomials

令
$$M_j(z) = \prod_{i\in C_j}(z - \alpha^i).$$
$C_j$ 在乘以 $2$ 之下封閉，所以 $M_j \in \mathbb F_2[z]$。  
它就是 1005 講義中 $\alpha^j$ 在 $\mathbb F_2$ 上的 minimal polynomial：  
若 $f \in \mathbb F_2[z]$ 且 $f(\alpha^j) = 0$，則 $f(\alpha^{2j}) = f(\alpha^j)^2 = 0$。  
重複下去，$M_j$ 的每個根都是 $f$ 的根，而這些根互不相同，所以 $M_j$ 整除 $f$。

取每個 cyclotomic coset 中最小的元素當代表，令 $J$ 是這些代表的集合，  
例如 $n = 15$ 時 $J = \{0, 1, 3, 5, 7\}$。  
每個 $\alpha^i$ 恰好是一個 $M_j$（$j \in J$）的根，因此
$$z^n - 1 = \prod_{j\in J} M_j(z)$$
就是 $z^n - 1$ 在 $\mathbb F_2[z]$ 中的 irreducible factorization，其中 $\deg M_j = |C_j|$。  
Binary cyclic codes 的 generator polynomials 就是其中一部分 $M_j$ 的乘積，  
共有 $2^{|J|}$ 個 binary cyclic codes。

### 例子：$z^7 - 1$

用 1005 講義的 $\mathbb F_8$：$\alpha^3 = \alpha + 1$。  
1005 講義已經看過 $\alpha \mapsto \alpha^2 \mapsto \alpha^4 \mapsto \alpha$，  
這三個元素都是 $x^3 + x + 1$ 的根，所以 $M_1(z) = z^3 + z + 1$。  
另一個 coset 的 $\beta = \alpha^3 = \alpha + 1$ 滿足
$$\beta^2 = \alpha^2 + 1,\qquad \beta^3 = (\alpha + 1)(\alpha^2 + 1) = \alpha^3 + \alpha^2 + \alpha + 1 = \alpha^2.$$
因此 $\beta^3 + \beta^2 + 1 = 0$，$M_3(z) = z^3 + z^2 + 1$。

所以
$$z^7 - 1 = (z + 1)(z^3 + z + 1)(z^3 + z^2 + 1).$$
三個 irreducible factors 給出 $2^3 = 8$ 個 binary cyclic codes：

| $g(z)$ | Zeros | 參數 | 名稱 |
|---|---|---|---|
| $1$ | 無 | $[7, 7, 1]$ | $\mathbb F_2^7$ |
| $z + 1$ | $C_0$ | $[7, 6, 2]$ | Single parity-check |
| $z^3 + z + 1$ | $C_1$ | $[7, 4, 3]$ | Hamming |
| $z^3 + z^2 + 1$ | $C_3$ | $[7, 4, 3]$ | Hamming（座標順序反過來） |
| $(z + 1)(z^3 + z + 1)$ | $C_0 \cup C_1$ | $[7, 3, 4]$ | Simplex，也是 Hamming 的 even-weight subcode |
| $(z + 1)(z^3 + z^2 + 1)$ | $C_0 \cup C_3$ | $[7, 3, 4]$ | Simplex |
| $(z^3 + z + 1)(z^3 + z^2 + 1)$ | $C_1 \cup C_3$ | $[7, 1, 7]$ | Repetition |
| $z^7 - 1$ | 全部 | $[7, 0]$ | 只有零向量 |

例如 zeros 為 $C_1$ 的 code，條件是 $c(\alpha) = \sum_i c_i \alpha^i = 0$。  
把 $\alpha^i$ 寫成 $\mathbb F_2$ 上的 $3$ 維向量，這個條件就是 $cH = 0$，其中 $H$ 的第 $i$ row 是 $\alpha^i$ 的係數。  
$H$ 的 $7$ 個 rows 恰好是 $\mathbb F_2^3$ 的全部非零向量，只是順序和 0914 講義不同。  
所以這正是 Hamming code，而且在這個座標順序下是 cyclic；同樣的論證對任意 $m$ 都成立。

### 例子：$z^{15} - 1$

用 primitive polynomial $x^4 + x + 1$ 建構 $\mathbb F_{16}$，令 $\alpha$ 是 $x$ 的餘數類，$\alpha^4 = \alpha + 1$。  
下表把 $\alpha^i$ 寫成四個 bits，從左到右是 $\alpha^3, \alpha^2, \alpha, 1$ 的係數；加法就是 XOR。

| $i$ | $\alpha^i$ | $i$ | $\alpha^i$ | $i$ | $\alpha^i$ |
|---|---|---|---|---|---|
| $0$ | `0001` | $5$ | `0110` | $10$ | `0111` |
| $1$ | `0010` | $6$ | `1100` | $11$ | `1110` |
| $2$ | `0100` | $7$ | `1011` | $12$ | `1111` |
| $3$ | `1000` | $8$ | `0101` | $13$ | `1101` |
| $4$ | `0011` | $9$ | `1010` | $14$ | `1001` |

五個 cyclotomic cosets 給出
$$M_0 = z + 1,\qquad M_1 = z^4 + z + 1,\qquad M_3 = z^4 + z^3 + z^2 + z + 1,$$
$$M_5 = z^2 + z + 1,\qquad M_7 = z^4 + z^3 + 1.$$
次數加起來是 $1 + 4 + 4 + 2 + 4 = 15$，所以共有 $2^5 = 32$ 個長度 $15$ 的 binary cyclic codes。  
$M_5$ 的兩個根 $\alpha^5, \alpha^{10}$ 的 order 是 $3$，它們和 $0, 1$ 組成 $\mathbb F_{16}$ 中的子域 $\mathbb F_4$。

下面用 SageMath 直接計算每個 $\prod_{i\in C_j}(z - \alpha^i)$。  
乘積在 $\mathbb F_{16}[z]$ 中計算，但結果的係數都是 $0$ 或 $1$。  
第一行輸出用整數 $a_0 + 2a_1 + 4a_2 + 8a_3$ 標記 $\alpha^i$，例如 $\alpha^7$ 是 $11$，也就是 `1011`。  
沿用 1012 講義，一律用 `**` 表示次方。

```python
W = PolynomialRing(GF(2), "w")
w = W.gen()
K = GF(2**4, "alpha", modulus=w**4 + w + 1)
alpha = K.gen()
Kz = PolynomialRing(K, "z")
z = Kz.gen()
n = 15
print([(alpha**i).to_integer() for i in range(n)])
```

```python
cosets = []
for j in range(n):
    if not any(j in C for C in cosets):
        cosets.append(sorted({j * 2**i % n for i in range(4)}))
for C in cosets:
    print(C, prod(z - alpha**i for i in C))
```

## BCH Bound

### 連續的零點

Cyclic code 的 zeros 可以直接給出 minimum distance 的下界。  
回想 0914 講義：若 $\mathcal C = \{c : cH = 0\}$，  
則 weight 為 $w$ 的非零 codeword 給出 $H$ 的某 $w$ 個 rows 之間的線性相依。  
所以若 $H$ 的任意 $\delta - 1$ 個 rows 都線性獨立，則 $d(\mathcal C) \ge \delta$。

**定理（BCH bound）：** 令 $\mathcal C$ 是長度 $n = 2^m - 1$ 的 binary cyclic code。  
若存在 $b$ 使
$$\alpha^b, \alpha^{b+1}, \ldots, \alpha^{b+\delta-2}$$
都是 $\mathcal C$ 的 zeros（共 $\delta - 1$ 個連續的冪次，指數取模 $n$），則 $d(\mathcal C) \ge \delta$。

證明：令 $H$ 是 $n \times (\delta - 1)$ 的矩陣，  
row 編號 $i = 0, \ldots, n - 1$，column 編號 $j = b, \ldots, b + \delta - 2$：
$$H = \bigl(\alpha^{ij}\bigr),\qquad cH = \bigl(c(\alpha^b), c(\alpha^{b+1}), \ldots, c(\alpha^{b+\delta-2})\bigr).$$
每個 codeword 都滿足 $cH = 0$。  
任取 $H$ 的 $\delta - 1$ 個 rows，編號為 $i_1, \ldots, i_{\delta-1}$，令 $X_l = \alpha^{i_l}$。

這些 rows 組成的方陣是
$$\begin{pmatrix}
X_1^b & X_1^{b+1} & \cdots & X_1^{b+\delta-2} \\
\vdots & \vdots & & \vdots \\
X_{\delta-1}^b & X_{\delta-1}^{b+1} & \cdots & X_{\delta-1}^{b+\delta-2}
\end{pmatrix}
= \operatorname{diag}(X_1^b, \ldots, X_{\delta-1}^b) V^\top,$$
其中 $V = V(X_1, \ldots, X_{\delta-1})$ 是 1012 講義的 Vandermonde matrix。  
$X_l$ 互不相同且不為零，所以行列式不為零，這些 rows 在 $\mathbb F_{2^m}$ 上線性獨立，在 $\mathbb F_2$ 上當然也是。  
因此非零 codeword 的 weight 至少是 $\delta$。

證明沒有用到 codeword 是 binary，只用到 $\alpha^0, \ldots, \alpha^{n-1}$ 互不相同。  
1012 講義的 cyclic RS code 有 zeros $\alpha, \ldots, \alpha^{n-k}$，共 $n - k$ 個連續冪次，  
所以 BCH bound 給出 $d \ge n - k + 1$，和 root bound 的結論相同。

### 例子：Hamming codes 的 $d \ge 3$

前面看到 zeros 為 $C_1$ 的 binary cyclic code 就是 Hamming code。  
$C_1 = \{1, 2, 4, \ldots\}$ 包含連續的 $1, 2$，所以 $d \ge 3$。  
Zeros 為 $C_0 \cup C_1$ 時包含連續的 $0, 1, 2$，所以 Hamming code 的 even-weight subcode 有 $d \ge 4$。  
$n = 7$ 的 $C_3 = \{3, 6, 5\}$ 包含連續的 $5, 6$，所以另一個 Hamming code 也有 $d \ge 3$。

## Binary BCH Codes

### 定義

令 $n = 2^m - 1$、$2 \le \delta \le n$。  
Designed distance 為 $\delta$ 的 **binary BCH code** 是 zeros 恰好為
$$C_1 \cup C_2 \cup \cdots \cup C_{\delta-1}$$
的 binary cyclic code，也就是包含 zeros $\alpha, \alpha^2, \ldots, \alpha^{\delta-1}$ 的 binary cyclic codes 中最大的一個。  
它的 generator polynomial 是
$$g(z) = \operatorname{lcm}\bigl(M_1(z), M_2(z), \ldots, M_{\delta-1}(z)\bigr),$$
記為 $\mathrm{BCH}(n, \delta)$。  
由 BCH bound，$d \ge \delta$。

Hocquenghem 在 1959 年提出這類 codes，Bose 與 Ray-Chaudhuri 在 1960 年獨立提出，BCH 就是三人姓氏的縮寫。  
更一般的 BCH code 可以從 $\alpha^b$ 開始取連續冪次，也可以用 order 為 $n$ 的元素取代 primitive element；  
今天只看上面這種 **narrow-sense primitive** 的情形。

### Dimension：偶數次方免費

$C_{2j} = C_j$，所以 $M_{2j} = M_j$：  
只要 $\alpha^j$ 是 zero，$\alpha^{2j}$ 自動也是。  
取 $\delta = 2t + 1$。  
每個 $j \le 2t$ 都能寫成 $2^s j'$，其中 $j'$ 是奇數，而 $M_j = M_{j'}$，因此
$$g(z) = \operatorname{lcm}\bigl(M_1(z), M_3(z), \ldots, M_{2t-1}(z)\bigr).$$

每個 $M_j$ 的次數 $|C_j| \le m$，而 lcm 中至多有 $t$ 個不同的因式，所以
$$n - k = \deg g \le mt,\qquad k \ge n - mt.$$
要更正 $t$ 個 errors，binary BCH code 只需要至多 $mt = t\log_2(n + 1)$ 個 parity bits。  
同樣的原因，$\delta = 2t$ 與 $\delta = 2t + 1$ 總是給出同一個 binary BCH code：$M_{2t} = M_t$ 已經在 lcm 中。

### 和 sphere packing bound 比較

0907 講義的 sphere packing：能更正 $t$ 個 errors 時，以 codewords 為中心、半徑 $t$ 的球互不相交，  
所以 $[n, k]$ binary linear code 滿足
$$2^k \sum_{i=0}^t \binom ni \le 2^n.$$
$t = 2$、$n = 2^m - 1$ 時，
$$1 + n + \binom n2 = 2^{2m-1} - 2^{m-1} + 1 > 2^{2m-2},$$
所以任何能更正 $2$ 個 errors 的 binary linear code 都需要 $n - k \ge 2m - 1$。  
$m \ge 3$ 時 $|C_3| = m$，$\mathrm{BCH}(n, 5)$ 恰好用 $2m$ 個 parity bits，只比這個下界多 $1$ 個。

一般而言，$t$ 固定、$n$ 很大時，$\log_2\binom nt \approx mt - \log_2 t!$，  
所以 sphere packing bound 要求的 parity bits 約為 $mt - \log_2 t!$，而 BCH 用至多 $mt$ 個。  
對小的 $t$，binary BCH codes 幾乎是最好的 linear codes。

### 例子：長度 $15$

$n = 15$ 時，各 designed distance 給出的 BCH codes：

| $\delta$ | $g(z)$ | Zeros | 參數 |
|---|---|---|---|
| $2, 3$ | $M_1 = z^4 + z + 1$ | $C_1$ | $[15, 11, 3]$ Hamming |
| $4, 5$ | $M_1 M_3 = z^8 + z^7 + z^6 + z^4 + 1$ | $C_1 \cup C_3$ | $[15, 7, 5]$ |
| $6, 7$ | $M_1 M_3 M_5 = z^{10} + z^8 + z^5 + z^4 + z^2 + z + 1$ | $C_1 \cup C_3 \cup C_5$ | $[15, 5, 7]$ |
| $8$ 至 $15$ | $M_1 M_3 M_5 M_7 = 1 + z + \cdots + z^{14}$ | $C_1 \cup C_3 \cup C_5 \cup C_7$ | $[15, 1, 15]$ repetition |

下面列出 $[15, 7, 5]$ 全部 $2^7 = 128$ 個 codewords 的 weights，確認 minimum distance 恰好是 $5$。  
`P` 是 $\mathbb F_2[z]$；SageMath 的 `minpoly()` 直接給出 minimal polynomial。

```python
from collections import Counter
from itertools import product

P = PolynomialRing(GF(2), "z")
M = {j: P((alpha**j).minpoly().list()) for j in [0, 1, 3, 5, 7]}
g = M[1] * M[3]
k = n - g.degree()
codewords = [vector(GF(2), (P(list(u)) * g).padded_list(n)) for u in product(GF(2), repeat=k)]
print(g, k)
print(sorted(Counter(c.hamming_weight() for c in codewords).items()))
```

$[15, 5, 7]$ 的非零 weights 只有 $7$、$8$、$15$，和刪掉一個座標的 $\mathrm{RM}(1, 4)$ 相同。  
事實上它就是 1005 週五作業的 $\mathrm{RM}(1, 4)^*$：  
在 $1, \alpha, \ldots, \alpha^{14}$ 的座標順序下，兩者是同一個 code。

### Designed distance 與真正的 minimum distance

Designed distance 只保證 $d \ge \delta$，真正的 minimum distance 可能更大。  
第一種原因：加入 $M_j$ 時，同一個 coset 的其他根也一起成為 zeros，可能湊出更長的連續冪次。  
例如 $n = 31$、$\delta = 8$ 時，zeros 包含 $C_1, C_3, C_5, C_7$，  
其中 $C_5 = \{5, 10, 20, 9, 18\}$ 補上了 $9$ 與 $10$，所以 zeros 包含 $\alpha^1, \ldots, \alpha^{10}$，其實 $d \ge 11$（van Lint §6.6 的例子）。  
前面 $n = 15$、$\delta = 8$ 的情形更極端：真正的 minimum distance 是 $15$。

第二種原因：即使連續冪次已經用完，BCH bound 也可能不 tight。  
例如 $n = 23$：$2^{11} = 2048 = 89 \cdot 23 + 1$，所以 $\mathbb F_{2^{11}}$ 中有 order 為 $23$ 的元素 $\beta$。  
用 $\beta$ 取代 $\alpha$，同樣的理論給出 cyclotomic coset
$$C_1 = \{1, 2, 4, 8, 16, 9, 18, 13, 3, 6, 12\}.$$
它包含連續的 $1, 2, 3, 4$，BCH bound 給出 $d \ge 5$。  
但這個 $[23, 12]$ code 的真正 minimum distance 是 $7$，它就是著名的 **binary Golay code**。

```python
L = GF(2**11, "b")
beta = L.multiplicative_generator()**89
golay = P(beta.minpoly().list())
print(beta.multiplicative_order(), golay)
weights = Counter((P(list(u)) * golay).hamming_weight() for u in product(GF(2), repeat=12))
print(sorted(weights.items()))
```

反方向也有保證：van Lint 證明了 primitive binary BCH code 的 minimum distance 至多是 $2\delta - 1$（Corollary 6.6.14）。  
所以 designed distance 至少抓到真正 minimum distance 的一半。

## Subfield Subcodes

### BCH code 是 RS code 的 binary 部分

在 $\mathbb F_{2^m}$ 上，取 1012 講義的 cyclic RS code，令 $n - k = \delta - 1$：
$$\mathrm{RS} = \{c \in \mathbb F_{2^m}^n : c(\alpha^j) = 0,\qquad j = 1, \ldots, \delta - 1\}.$$
對 binary 向量 $c$，$c(\alpha^j) = 0$ 當且僅當 $M_j$ 整除 $c(z)$（minimal polynomial 的性質），因此
$$\boxed{\mathrm{BCH}(n, \delta) = \mathrm{RS} \cap \mathbb F_2^n.}$$
大的域上的 code 中，分量全部落在子域的那些 codewords 形成的 code，稱為 **subfield subcode**。

Subfield subcode 的 codewords 只是原本 codewords 的一部分，所以 minimum distance 不會變小：
$$d\bigl(\mathrm{BCH}(n, \delta)\bigr) \ge d(\mathrm{RS}) = \delta.$$
這是 BCH bound（$b = 1$ 的情形）的另一個證明。  
代價是 dimension：$\mathrm{RS}$ 有 $(2^m)^{n-\delta+1}$ 個 codewords，其中只有很少數是 binary。

例如 $n = 15$、$\delta = 5$：  
RS code 是 $[15, 11, 5]_{16}$，有 $16^{11} = 2^{44}$ 個 codewords；  
其中 binary 的只有 $2^7 = 128$ 個，形成 $[15, 7, 5]$ BCH code。  
兩者的 minimum distance 都是 $5$，但 dimension 從 $11$ 掉到 $7$。

### 展開成 binary parity-check matrix

$c$ 是 binary 時，$c(\alpha^{2j}) = c(\alpha^j)^2$，所以只需要奇數次方的條件。  
$\delta = 2t + 1$ 時，
$$c \in \mathrm{BCH}(n, \delta)
\quad\Longleftrightarrow\quad
c(\alpha^j) = 0\quad\text{for }j = 1, 3, \ldots, 2t - 1.$$
把每個 $\alpha^{ij}$ 寫成 $m$ 個 bits，得到 $n \times mt$ 的 binary 矩陣 $H_2$：  
它的第 $i$ row 是 $\alpha^i, \alpha^{3i}, \ldots, \alpha^{(2t-1)i}$ 的 bits 接起來。  
對 binary 的 $c$，$c(\alpha^j) = 0$ 就是 $cH_2$ 中對應的 $m$ 個 bits 都為零。

$t = 1$ 時，$H_2$ 的 rows 就是 $\alpha^i$ 的 bits，也就是前面 Hamming code 的 parity-check matrix。  
$t = 2$、$n = 15$ 時，$H_2$ 是 $15 \times 8$ 矩陣，rank 是 $8 = n - k$。  
但 $H_2$ 的 columns 不一定線性獨立：  
例如 $\delta = 7$ 時 $M_5$ 只有 $2$ 次，$\alpha^{5i}$ 的 $4$ 個 bits 只貢獻 rank $2$。

```python
def to_bits(a):
    return a.polynomial().padded_list(4)

H2 = matrix(GF(2), [to_bits(alpha**i) + to_bits(alpha**(3*i)) for i in range(n)])
print(H2.rank())
print(all(c * H2 == 0 for c in codewords))
```

## Decoding Binary BCH Codes

### Syndromes 只需要算一半

傳送 $c$，收到 $y = c + e$，其中 $e \in \mathbb F_2^n$。  
和 1012 講義相同，若錯誤位置是 $i_1, \ldots, i_\nu$，令 $X_l = \alpha^{i_l}$，則 syndromes 是
$$S_j = y(\alpha^j) = e(\alpha^j) = \sum_{l=1}^\nu X_l^j,\qquad j = 1, \ldots, 2t.$$
和 RS 不同的是，錯誤值 $Y_l$ 全部是 $1$。

兩個簡化：
1. $S_{2j} = e(\alpha^{2j}) = e(\alpha^j)^2 = S_j^2$，所以只需要計算 $S_1, S_3, \ldots, S_{2t-1}$。
2. 找到錯誤位置後直接翻轉那些 bits，不需要 Forney formula。

Peterson's method、Chien search 都和 1012 講義相同。

### 更正兩個 errors 的公式

取 $t = 2$。  
Characteristic 為 $2$，減號就是加號，所以 error locator polynomial 是
$$\Lambda(z) = (1 + X_1 z)(1 + X_2 z) = 1 + \Lambda_1 z + \Lambda_2 z^2,\qquad \Lambda_1 = X_1 + X_2,\quad \Lambda_2 = X_1 X_2.$$
與其解 Peterson 的線性方程組，不如直接從 $S_1, S_3$ 解出 $\Lambda_1, \Lambda_2$。

首先 $S_1 = X_1 + X_2 = \Lambda_1$。  
其次
$$S_3 = X_1^3 + X_2^3 = (X_1 + X_2)(X_1^2 + X_1 X_2 + X_2^2) = S_1(S_1^2 + \Lambda_2),$$
因為 $X_1^2 + X_2^2 = (X_1 + X_2)^2 = S_1^2$。  
所以 $S_1 \ne 0$ 時
$$\Lambda_1 = S_1,\qquad \Lambda_2 = \frac{S_3 + S_1^3}{S_1}.$$
這和 1012 講義的 $\det M_2 = S_1 S_3 - S_2^2 = S_1(S_3 + S_1^3)$ 一致。

至多兩個 errors 時，decoder 依照 $S_1, S_3$ 分成四種情形：

| 條件 | 判斷 |
|---|---|
| $S_1 = 0$、$S_3 = 0$ | 沒有 errors |
| $S_1 \ne 0$、$S_3 = S_1^3$ | 一個 error，error locator 是 $X_1 = S_1$ |
| $S_1 \ne 0$、$S_3 \ne S_1^3$ | 兩個 errors，$\Lambda(z)$ 的兩個根給出位置 |
| $S_1 = 0$、$S_3 \ne 0$ | 至少三個 errors，宣告失敗 |

兩個不同的 $X_1, X_2$ 相加不會是零，所以第四種情形不可能來自至多兩個 errors。  
第三種情形若 $\Lambda(z)$ 沒有兩個不同的根 $\alpha^{-i}$，同樣表示至少三個 errors。

### 例子：$[15, 7, 5]$ 更正兩個 errors

取最簡單的非零 codeword：$g(z)$ 本身，
$$c(z) = 1 + z^4 + z^6 + z^7 + z^8.$$
在座標 $3$ 與 $10$ 翻轉 bits，收到
$$y(z) = 1 + z^3 + z^4 + z^6 + z^7 + z^8 + z^{10}.$$
Decoder 只知道 $y$。

用冪次表的 bits 計算 $S_1 = y(\alpha)$，加法就是 XOR：
$$\begin{aligned}
S_1 &= \alpha^0 + \alpha^3 + \alpha^4 + \alpha^6 + \alpha^7 + \alpha^8 + \alpha^{10} \\
&= \texttt{0001} + \texttt{1000} + \texttt{0011} + \texttt{1100} + \texttt{1011} + \texttt{0101} + \texttt{0111} \\
&= \texttt{1111} = \alpha^{12}.
\end{aligned}$$
$S_3 = y(\alpha^3)$ 的各項是 $\alpha^{3i}$，指數取模 $15$，相同的項兩兩抵消：
$$S_3 = \alpha^0 + \alpha^9 + \alpha^{12} + \alpha^3 + \alpha^6 + \alpha^9 + \alpha^0 = \alpha^{12} + \alpha^3 + \alpha^6 = \texttt{1011} = \alpha^7.$$

$S_1^3 = \alpha^{36} = \alpha^6$，與 $S_3 = \alpha^7$ 不同，所以有兩個 errors：
$$\Lambda_2 = \frac{S_3 + S_1^3}{S_1} = \frac{\alpha^7 + \alpha^6}{\alpha^{12}} = \frac{\alpha^{10}}{\alpha^{12}} = \alpha^{13}.$$
因此
$$\Lambda(z) = 1 + \alpha^{12} z + \alpha^{13} z^2.$$

Chien search：逐一代入 $z = \alpha^{-i}$。  
例如 $i = 3$ 時 $\alpha^{-3} = \alpha^{12}$，
$$\Lambda(\alpha^{12}) = 1 + \alpha^{24} + \alpha^{37} = 1 + \alpha^9 + \alpha^7 = \texttt{0001} + \texttt{1010} + \texttt{1011} = 0.$$
同樣得到 $i = 10$ 也是根，所以錯誤位置是 $3$ 與 $10$，翻轉這兩個 bits 就得到 $c$。  
驗證：$X_1 + X_2 = \alpha^3 + \alpha^{10} = \texttt{1000} + \texttt{0111} = \alpha^{12} = \Lambda_1$，而 $X_1 X_2 = \alpha^{13} = \Lambda_2$。

```python
y = P([1, 0, 0, 1, 1, 0, 1, 1, 1, 0, 1])
S1, S3 = y(alpha), y(alpha**3)
Lambda2 = (S3 + S1**3) / S1
Lambda = Kz([1, S1, Lambda2])
print(S1 == alpha**12, S3 == alpha**7, Lambda2 == alpha**13)
print("error positions:", [i for i in range(n) if Lambda(alpha**(-i)) == 0])
```

### 寫成 decoder

把四種情形寫成 function，輸出更正後的 codeword，失敗時回傳 `None`。  
接著對 $[15, 7, 5]$ 的一個 codeword，試遍全部 $1 + 15 + \binom{15}2 = 121$ 種至多兩個 errors 的 patterns。

```python
def bch2_decode(y):
    y = vector(GF(2), y)
    S1, S3 = P(list(y))(alpha), P(list(y))(alpha**3)
    if S1 == 0:
        return y if S3 == 0 else None
    Lambda = Kz([1, S1, (S3 + S1**3) / S1])
    positions = [i for i in range(n) if Lambda(alpha**(-i)) == 0]
    if len(positions) != Lambda.degree():
        return None
    for i in positions:
        y[i] += 1
    return y
```

```python
from itertools import combinations

c = codewords[37]
successes = 0
for r in range(3):
    for positions in combinations(range(n), r):
        e = vector(GF(2), n)
        for i in positions:
            e[i] = 1
        successes += bch2_decode(c + e) == c
print(successes)
```

## 應用

### QR code 的 format information

1012 講義提到 QR code 的資料用 $\mathbb F_{256}$ 上的 RS codes 保護。  
但掃描器必須先知道 error correction level 與 mask pattern，才能開始讀資料。  
這 $5$ 個 bits 稱為 **format information**，用 $[15, 5, 7]$ BCH code 保護，  
generator polynomial $g(z)$ 正是前面的 $M_1 M_3 M_5 = z^{10} + z^8 + z^5 + z^4 + z^2 + z + 1$。

QR code 規格從最高次項開始寫 bits。  
例如 level M（`00`）、mask `101` 的資訊 bits 是 `00101`，對應 $m(z) = z^2 + 1$。  
和 1012 講義的 systematic encoding 一樣，計算 $z^{10} m(z)$ 除以 $g(z)$ 的餘數：
$$r(z) = z^7 + z^6 + z^4 + z^3 + z^2.$$
得到 codeword `001010011011100`：前 $5$ 個 bits 是資訊，後 $10$ 個是 $r(z)$ 從 $z^9$ 到 $z^0$ 的係數。

最後再和固定的 mask `101010000010010` 做 XOR，得到 `100000011001110`，避免 format information 全部是 $0$。  
XOR 一個固定向量不改變兩個字串之間的距離，所以仍然能更正 $3$ 個 errors。  
合法的字串只有 $2^5 = 32$ 個，掃描器可以直接和全部 $32$ 個比較，挑最近的一個。  
同一份 format information 在 QR code 上印兩次，放在不同的位置。

```python
g7 = M[1] * M[3] * M[5]
m = P([1, 0, 1])
r = m.shift(10) % g7
word = "".join(str(b) for b in reversed((m.shift(10) + r).padded_list(15)))
mask = "101010000010010"
print(g7)
print(word)
print("".join(str(int(a != b)) for a, b in zip(word, mask)))
```

### NAND flash 與 DVB-S2

NAND flash 的 cell 用電荷存 bits，電荷會隨著時間與讀寫次數漂移，讀出時出現零星的 bit errors。  
NAND flash controller 長期使用 binary BCH codes，最近的 SSD 則多半改用 LDPC codes。  
例如把每 $512$ bytes，也就是 $4{,}096$ bits 的資料當作一組，  
需要 $n \ge 4{,}096 + mt$，所以取 $m = 13$（$n \le 8{,}191$），再 shorten 到需要的長度。  
更正 $t = 8$ 個 bit errors 只需要 $13 \cdot 8 = 104$ bits，也就是 $13$ bytes 的 parity。  
Linux kernel 的 `lib/bch.c` 就是這類 binary BCH encoder 與 decoder。

衛星電視的 DVB-S2 標準把兩層 codes 串接：內層是很長的 LDPC code，外層是 binary BCH code。  
LDPC decoder 偶爾會留下少量的 bit errors（error floor），外層的 BCH 每個 frame 更正 $t \in \{8, 10, 12\}$ 個 bits 來清理它們。  
這和 1012 講義的 CCSDS 一樣，是「內層對抗通道、外層清理殘留錯誤」的串接結構。

## 延伸話題

### Berlekamp–Massey algorithm

$t$ 大的時候，Peterson's method 要反覆解線性方程組，太慢。  
1012 講義提過的 Berlekamp–Massey algorithm 把 syndromes 看成一個數列，  
逐項找出能產生 $S_1, \ldots, S_j$ 的最短 linear recurrence，也就是最短的 LFSR（linear feedback shift register）。  
每多讀一個 syndrome，只在預測錯誤時用之前的紀錄修正 $\Lambda$，總共只需要 $O(t^2)$ 次有限體運算。  
Binary 的情形還可以跳過一半的步驟：因為 $S_{2j} = S_j^2$，偶數步的預測一定正確。

### 長的 BCH codes

$t$ 固定時，binary BCH codes 非常接近 sphere packing bound。  
但若讓 $n \to \infty$，同時要求 rate $k/n$ 與相對距離 $d/n$ 都不趨近 $0$，primitive BCH codes 做不到：  
van Lint §6.6 指出，長的 primitive BCH codes 一定有 $k/n \to 0$ 或 $d/n \to 0$。  
兩者都不趨近 $0$ 的 code family 稱為 **asymptotically good**；  
第一個 explicit construction 是 Justesen（1972），見 van Lint 第 11 章。

### 非線性的 codes 可以更好

$[15, 7, 5]$ 有 $128$ 個 codewords。  
Sphere packing bound 說長度 $15$、$d = 5$ 的 binary code 至多有 $\lfloor 2^{15}/121\rfloor = 270$ 個 codewords。  
事實上存在 $256$ 個 codewords 的非線性 code：把 Nordstrom–Robinson code（長度 $16$、$256$ 個 codewords、$d = 6$）刪掉一個座標。  
同樣大小的 linear code 會是 $[15, 8, 5]$，加上 overall parity bit 就變成 $[16, 8, 6]$，而後者並不存在。  
van Lint 第 8 章用 $\mathbb Z_4$ 上的 linear codes 解釋這種非線性 binary codes。
