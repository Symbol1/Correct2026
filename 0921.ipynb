## Block Codes 與 Linear Codes


### Block code 的數學定義


給定一個有限 alphabet $\mathcal A$  
令 $|\mathcal A| = q$  
長度為 $n$ 的 **block code** 就是一個非空集合  
$$C \subseteq \mathcal A^n.$$
集合裡的元素稱為 codewords  
所有 codewords 的長度都相同  

若 $|C| = M$、minimum distance 是 $d$  
記為 $(n, M, d)_q$ code


編碼器把不同訊息對應到不同 codewords  
$$\operatorname{Enc} : \mathcal M \longrightarrow C.$$
通常選擇一個 bijection，讓 $|\mathcal M| = |C|$  
code 是「哪些字串合法」的集合  
同一個 code 可以搭配不同的訊息標籤與編碼器  

用 $q$-ary symbols 計算的碼率是  
$$R = \frac{\log_q M}{n}.$$
一般 block code 不要求 $M$ 是 $q$ 的整數次方


### Linear code 的數學定義


令 $q$ 是 prime power，使用有限域 $\mathbb F_q$  
若 $C$ 是 $\mathbb F_q^n$ 的線性子空間  
就稱為 **linear code**  
也就是包含零向量，且對線性組合封閉  
$$u, v \in C,\quad \alpha, \beta \in \mathbb F_q
\quad\Longrightarrow\quad \alpha u + \beta v \in C.$$
若 $\dim C = k$  
則 $|C| = q^k$，碼率為 $k/n$  
參數記為 $[n, k, d]_q$；binary 時常省略下標 2


### Generator 與 parity-check matrix


選 $C$ 的一組基底，排成 $k \times n$ 的矩陣 $G$  
就有  
$$C = \{aG : a \in \mathbb F_q^k\}.$$
延續本講義的 row-vector 慣例  
parity-check matrix $H$ 的大小是 $n \times (n - k)$  
要求 $H$ 有 rank $n - k$，且  
$$C = \{v \in \mathbb F_q^n : vH = 0\}.$$
因此 $GH = 0$  
兩個矩陣分別描述如何產生 codewords 與如何檢查合法性


### Dual code


與 $C$ 中每個向量都正交的向量，組成 dual code  
$$C^\perp = \{y \in \mathbb F_q^n : c \cdot y = 0\text{ 對所有 }c \in C\}.$$
其 dimension 是 $n - k$  
依本講義的矩陣慣例，$H^{\mathsf T}$ 是 $C^\perp$ 的 generator matrix  

所以先前從 Hamming 的 $H^{\mathsf T}$ 得到的 simplex code  
正是 Hamming code 的 dual  
同樣地，extended Hamming 的 dual 是 $\mathrm{RM}(1, m)$


## Minimum Distance 與 Minimum Weight


### Hamming distance


兩個長度為 $n$ 的向量 $u, v$  
其 **Hamming distance** 是不同座標的數量  
$$d_H(u, v) = |\{i : u_i \ne v_i\}|.$$
例如  
$$d_H(0101, 1100) = 2.$$
當 $|C| \ge 2$  
code 的 minimum distance 定義為  
$$d(C) = \min_{u, v \in C,\ u \ne v} d_H(u, v).$$


### Hamming weight


對 $v \in \mathbb F_q^n$  
其 **Hamming weight** 是非零座標的數量  
$$\operatorname{wt}(v) = |\{i : v_i \ne 0\}| = d_H(v, 0).$$
在 binary 的情況下，就是 1 的數量  
若 $C$ 含有非零向量，定義 minimum nonzero weight  
$$w_{\min}(C) = \min_{v \in C,\ v \ne 0} \operatorname{wt}(v).$$
需要排除零向量  
否則 linear code 的最小 weight 永遠是 0


### Linear code：$d(C) = w_{\min}(C)$


對任何 $u, v \in \mathbb F_q^n$  
都有  
$$d_H(u, v) = \operatorname{wt}(u - v).$$
若 $C$ 是非零 linear code  
不同 codewords 的差恰好遍歷 $C$ 的所有非零向量  
因此  
$$d(C) = \min_{u \ne v} \operatorname{wt}(u - v) = w_{\min}(C).$$
所以分析 linear code 時，只需要把其他 codewords 跟零向量比較


### 一般 block code 不一定成立


例如  
$$C = \{000, 110, 111\}.$$
非零 codewords 的 weights 是 2 與 3  
所以 $w_{\min}(C) = 2$  
但  
$$d_H(110, 111) = 1.$$
因此 $d(C) = 1$  
這個集合對加法不封閉，因為 $110 + 111 = 001 \notin C$


## Weight Enumerator 與 Distance Enumerator


### 把每種 weight 的數量記下來


令 $A_i$ 表示 weight 為 $i$ 的 codewords 數量  
$$A_i = |\{c \in C : \operatorname{wt}(c) = i\}|.$$
**Weight enumerator** 定義為  
$$W_C(z) = \sum_{i = 0}^n A_i z^i = \sum_{c \in C} z^{\operatorname{wt}(c)}.$$
其中 $z$ 是形式變數  
係數記錄數量，指數記錄 weight  

因此 $W_C(1) = |C|$  
對非零 linear code，最小正次方的指數就是 minimum distance


### Hadamard 與 RM 的例子


線性函數版本的 Hadamard code  
有一個零向量，其餘 $2^m - 1$ 個 codewords 都有一半的 1  
所以  
$$W_{\operatorname{Had}_m}(z) = 1 + (2^m - 1)z^{2^{m - 1}}.$$
加入常數項後，多了全一向量與原有 codewords 的補向量  
因此  
$$W_{\mathrm{RM}(1, m)}(z) = 1 + (2^{m + 1} - 2)z^{2^{m - 1}} + z^{2^m}.$$
例如 extended Hamming $[8, 4, 4]$ 的 weight enumerator 是  
$$1 + 14z^4 + z^8.$$


### 兩個變數的寫法


也可以同時記錄零與非零座標的數量  
$$W_C(X, Y) = \sum_{c \in C} X^{n - \operatorname{wt}(c)}Y^{\operatorname{wt}(c)}.$$
每一項的總次數都是 $n$  
代入 $X = 1$、$Y = z$，就回到 $W_C(z)$  

例如 extended Hamming $[8, 4, 4]$ 得到  
$$W_C(X, Y) = X^8 + 14X^4Y^4 + Y^8.$$


### Distance enumerator


Weight enumerator 是看 codewords 離零向量多遠  
Distance enumerator 則看 codewords 彼此多遠  

這裡採用「對每個起點取平均」的定義  
$$B_i = \frac{1}{|C|}|\{(u, v) \in C^2 : d_H(u, v) = i\}|.$$
配對 $(u, v)$ 有順序，也包含 $u = v$  
定義  
$$D_C(z) = \sum_{i = 0}^n B_i z^i = \frac{1}{|C|}\sum_{u, v \in C} z^{d_H(u, v)}.$$


這個 normalization 讓  
$$B_0 = 1,\qquad D_C(1) = |C|.$$
若 $|C| \ge 2$  
minimum distance 就是最小的 $i > 0$，使得 $B_i > 0$  

一般 code 的 $B_i$ 不一定是整數  
因為它是平均每個 codeword 有幾個距離為 $i$ 的鄰居  
若不除以 $|C|$，得到的有序配對計數多項式就是 $|C|D_C(z)$


### Linear code：兩種 enumerators 相同


固定 $u \in C$  
若 $C$ 是 linear code，映射 $v \mapsto v - u$ 是 $C$ 到自己的 bijection  
因此  
$$\sum_{v \in C} z^{d_H(u, v)} = \sum_{c \in C} z^{\operatorname{wt}(c)} = W_C(z).$$
每個起點看到的距離分布都一樣  
再對所有起點取平均  
$$\boxed{D_C(z) = W_C(z).}$$
minimum distance 等於 minimum nonzero weight，只是這個等式的一部分


### 同一個非線性例子


回到 $C = \{000, 110, 111\}$  
它的 weight enumerator 是  
$$W_C(z) = 1 + z^2 + z^3.$$
三個不同的無序配對，其距離分別是 1、2、3  
每個無序配對對應兩個有序配對  
加上三個自己到自己的配對  
$$D_C(z) = 1 + \frac{2}{3}z + \frac{2}{3}z^2 + \frac{2}{3}z^3.$$
Weight enumerator 看不到 codewords 之間距離為 1 的那一對  
Distance enumerator 則完整記錄了它
