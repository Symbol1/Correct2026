## Block Codes 與 Linear Codes


### Block code 的數學定義


給定一個有限字母集合 alphabet $\mathcal A$ ，記 $|\mathcal A| = q$  
一個長度 block lenth 為 $n$ 的 **block code** （有時候會說 **codebook**）  
是一個字串的集合  
$$\mathcal C \subseteq \mathcal A^n.$$
集合裡的元素稱為碼字 codewords

$q$-ary code of length $n$ 也是很常見的叫法

字母表大小 $q$ 常用以下英文名稱：

- $q = 2$：binary code（二元碼）
- $q = 3$：ternary code（三元碼）
- $q = 4$：quaternary code（四元碼）

其他 $q$ 的英文名稱可參考 Wikipedia  
[Numeral prefix](https://en.wikipedia.org/wiki/Numeral_prefix)。

若 $|\mathcal C| = M$  
令 $d$ 為任兩個不同 codewords 之間相異座標數量的最小值  
則稱為 $(n, M, d)_q$ code

注意到 $M$ 是大小 size， $n$ 是碼長

所有可能的原始訊息組成一個有限集合  
對於 $(n, M, d)_q$ code，取  
$$\mathcal M = [M] = \{1, \dotsc, M\},\qquad |\mathcal M| = M.$$
稱為訊息集合 message set，記作 $\mathcal M$  
其中每個 $m \in \mathcal M$ 代表一個可能的訊息 message  
這些元素只是訊息的標籤，一般不要求 $\mathcal M$ 具有代數結構


編碼器 encoder 是從訊息集合到 codebook 的 bijection  
$$\operatorname{Enc} : \mathcal M \longrightarrow \mathcal C.$$
因此每個訊息對應到唯一的 codeword  
且每個 codeword 都有對應的訊息  
此時 $|\mathcal M| = |\mathcal C| = M$  

用 $q$-ary symbols 計算的碼率是  
$$R = \frac{\log_q M}{n}.$$
一般 block code 不要求 $M$ 是 $q$ 的整數次方

### Linear code 的數學定義


令 $q$ 是 prime power，$\mathbb F_q$ 表示含有 $q$ 個元素的有限體  
若 $\mathcal C$ 是 $\mathbb F_q^n$ 的線性子空間，就稱為 **linear code**  
也就是包含零向量，並且滿足下面兩種封閉性  
對純量乘法封閉：  
$$u \in \mathcal C,\quad \alpha \in \mathbb F_q
\quad\Longrightarrow\quad \alpha u \in \mathcal C.$$
對加法封閉：  
$$u, v \in \mathcal C\quad\Longrightarrow\quad u + v \in \mathcal C.$$

u - v = u + (-1)v

若 $\dim \mathcal C = k$，則 $|\mathcal C| = q^k$，碼率為 $k/n$  
參數記為 $[n, k, d]_q$；binary 時常省略下標 2


### Generator matrix 與 parity-check matrix


選 $\mathcal C$ 的一組基底，排成 $k \times n$ 的矩陣 $G$  
$$\mathcal C = \{aG : a \in \mathbb F_q^k\}.$$
可以要求 G 是 full rank，  
也可以要求 G 長得很漂亮但不 full rank  
所以我不會問你「a generator matrix is full rank」這種是非題

parity-check matrix $H$ 指的是把所有 codeword 消滅的矩陣
$$\mathcal C = \{v \in \mathbb F_q^n : vH = 0\}.$$
因此 $GH = 0$

如果 $H$ 是 full rank，則它的大小是 $n \times (n - k)$  
但也可以不用是


### Dual code


對 $c, y \in \mathbb F_q^n$，定義 standard bilinear pairing  
$$c \cdot y = \sum_{i=1}^n c_i y_i.$$
這個 pairing 是 nondegenerate（非退化）的：  
若 $y \ne 0$，可選 $y_j \ne 0$；對第 $j$ 個標準基底向量 $e_j$，有 $e_j \cdot y = y_j \ne 0$  

與 $\mathcal C$ 中每個向量都正交的向量，組成 dual code  
$$\mathcal C^\bot = \{y \in \mathbb F_q^n : c \cdot y = 0 \text{ 對所有 } c \in \mathcal C\}.$$

![alt text](image.png)

若 $\dim \mathcal C = k$，令 $G$ 是 $k \times n$ 的 generator matrix  
則 $y \in \mathcal C^\bot$ 等價於 $Gy^\top = 0$  
因為 $\operatorname{rank}(G) = k$，rank-nullity theorem 給出  
$$\dim \mathcal C^\bot = n - k.$$

$H^\top$ 是 $\mathcal C^\bot$ 的 generator matrix  
$G^\top$ 是 $\mathcal C^\bot$ 的 parity check matrix

所以先前從 Hamming 的 $H^\top$ 得到的 simplex code  
正是 Hamming code 的 dual  
同樣地，extended Hamming 的 dual 是 $\mathrm{RM}(1, m)$

## Minimum Distance 與 Minimum Weight


### Hamming distance


兩個長度為 $n$ 的向量 $u, v$  
其 **Hamming distance** 是不同座標的數量  
$$d_H(u, v) = |\{i : u_i \ne v_i\}|.$$
例如  
$$d_H(0101, 1100) = 2.$$
當 $|\mathcal C| \ge 2$  
code 的 minimum distance 定義為  
$$d(\mathcal C) = \min_{u,v\in\mathcal C,u\ne v} d_H(u, v).$$


### Hamming weight


對 $v \in \mathbb F_q^n$  
其 **Hamming weight** 是非零座標的數量  
$$\operatorname{wt}(v) = |\{i : v_i \ne 0\}| = d_H(v, 0).$$
在 binary 的情況下，就是 1 的數量  
若 $\mathcal C$ 含有非零向量，定義 minimum nonzero weight  
$$w_{\min}(\mathcal C) = \min_{v\in\mathcal C,v\ne0} \operatorname{wt}(v).$$
需要排除零向量  
否則 linear code 的最小 weight 永遠是 0


### Linear code：$d(\mathcal C) = w_{\min}(\mathcal C)$


對任何 $u, v \in \mathbb F_q^n$  
都有  
$$d_H(u, v) = \operatorname{wt}(u - v).$$
若 $\mathcal C$ 是非零 linear code  
不同 codewords 的差恰好遍歷 $\mathcal C$ 的所有非零向量  
因此  
$$d(\mathcal C) = \min_{u\ne v} \operatorname{wt}(u - v) = w_{\min}(\mathcal C).$$
所以分析 linear code 時，只需要把其他 codewords 跟零向量比較


### 一般 block code 不一定成立


例如  
$$\mathcal C = \{000, 110, 111\}.$$
非零 codewords 的 weights 是 2 與 3  
所以 $w_{\min}(\mathcal C) = 2$  
但  
$$d_H(110, 111) = 1.$$
因此 $d(\mathcal C) = 1$  
這個集合對加法不封閉，因為 $110 + 111 = 001 \notin \mathcal C$


## Weight Enumerator 與 Distance Enumerator


### 把每種 weight 的數量記下來


令 $A_i$ 表示 weight 為 $i$ 的 codewords 數量  
$$A_i = |\{c \in \mathcal C : \operatorname{wt}(c) = i\}|.$$
令 $z$ 為形式變數，**weight enumerator** 定義為  
$$W_{\mathcal C}(z) = \sum_{i=0}^n A_i z^i = \sum_{c\in\mathcal C} z^{\operatorname{wt}(c)}.$$
可以把形式變數 $z$ 想成掛係數的衣架  
$z^i$ 標出 weight $i$ 的位置，讓係數 $A_i$ 掛在對應的位置上  
係數記錄數量，指數記錄 weight  

因此 $W_{\mathcal C}(1) = |\mathcal C|$  
對非零 linear code，最小正次方的指數就是 minimum distance


### Hadamard 與 RM 的例子


對 $m \ge 1$，將長度為 $2^m$ 的線性函數版本 Hadamard code 記為 $\operatorname{Had}_m$  
有一個零向量，其餘 $2^m - 1$ 個 codewords 都有一半的 1  
所以  
$$W_{\operatorname{Had}_m}(z) = 1 + (2^m - 1)z^{2^{m-1}}.$$

![alt text](image-1.png)

加入常數項後得到 $\mathrm{RM}(1, m)$，多了全一向量與原有 codewords 的補向量  
因此  
$$W_{\mathrm{RM}(1,m)}(z) = 1 + (2^{m+1} - 2)z^{2^{m-1}} + z^{2^m}.$$
例如 extended Hamming $[8, 4, 4]$ 的 weight enumerator 是  
$$1 + 14z^4 + z^8.$$


![alt text](image-2.png)

### 兩個變數的寫法


也可以同時記錄零與非零座標的數量  
令 $X, Y$ 為形式變數，定義  
$$W_{\mathcal C}(X, Y) = \sum_{c\in\mathcal C} X^{n-\operatorname{wt}(c)}Y^{\operatorname{wt}(c)}.$$
每一項的總次數都是 $n$  
代入 $X = 1$、$Y = z$，就回到 $W_{\mathcal C}(z)$  

例如 extended Hamming $[8, 4, 4]$ 得到  
$$W_{\mathcal C}(X, Y) = X^8 + 14X^4Y^4 + Y^8.$$


![alt text](image-3.png)

### Distance enumerator


Weight enumerator 是看 codewords 離零向量多遠  
Distance enumerator 則看 codewords 彼此多遠  

這裡採用「對每個起點取平均」的定義  
$$B_i = \frac{1}{|\mathcal C|}|\{(u, v) \in \mathcal C^2 : d_H(u, v) = i\}|.$$
配對 $(u, v)$ 有順序，也包含 $u = v$  
定義  
$$D_{\mathcal C}(z) = \sum_{i=0}^n B_i z^i = \frac{1}{|\mathcal C|}\sum_{u,v\in\mathcal C} z^{d_H(u,v)}.$$


這個 normalization 讓  
$$B_0 = 1,\qquad D_{\mathcal C}(1) = |\mathcal C|.$$
若 $|\mathcal C| \ge 2$  
minimum distance 就是最小的 $i > 0$，使得 $B_i > 0$  

一般 code 的 $B_i$ 不一定是整數  
因為它是平均每個 codeword 有幾個距離為 $i$ 的鄰居  
若不除以 $|\mathcal C|$，得到的有序配對計數多項式就是 $|\mathcal C|D_{\mathcal C}(z)$


### Linear code：兩種 enumerators 相同


固定 $u \in \mathcal C$  
若 $\mathcal C$ 是 linear code，映射 $v \mapsto v - u$ 是 $\mathcal C$ 到自己的 bijection  
因此  
$$\sum_{v\in\mathcal C} z^{d_H(u,v)} = \sum_{c\in\mathcal C} z^{\operatorname{wt}(c)} = W_{\mathcal C}(z).$$
每個起點看到的距離分布都一樣  
再對所有起點取平均  
$$\boxed{D_{\mathcal C}(z) = W_{\mathcal C}(z).}$$
minimum distance 等於 minimum nonzero weight，只是這個等式的一部分


### 同一個非線性例子


回到 $\mathcal C = \{000, 110, 111\}$  
它的 weight enumerator 是  
$$W_{\mathcal C}(z) = 1 + z^2 + z^3.$$
三個不同的無序配對，其距離分別是 1、2、3  
每個無序配對對應兩個有序配對  
加上三個自己到自己的配對  
$$D_{\mathcal C}(z) = 1 + \frac{2}{3}z + \frac{2}{3}z^2 + \frac{2}{3}z^3.$$
Weight enumerator 看不到 codewords 之間距離為 1 的那一對  
Distance enumerator 則完整記錄了它


### Weight enumerator 與 BEC、BSC

先考慮 $\operatorname{BEC}(\epsilon)$  
由 linearity，只需要分析傳送 all-zero codeword 的情形  
weight 為 $i$ 的 nonzero codeword 只有在其 nonzero positions 全被擦除時才無法排除，probability 為 $\epsilon^i$  
對所有 nonzero codewords 使用 union bound，得到  
$$P_{\mathrm{fail}}^{\mathrm{BEC}} \le \sum_{i=1}^n A_i \epsilon^i = W_{\mathcal C}(1, \epsilon) - 1.$$

![alt text](image-4.png)

接著考慮一般的 binary-input memoryless symmetric channel $W$  
它的 **Bhattacharyya parameter** 定義為  
$$Z(W) = \sum_y \sqrt{W(y \mid 0)W(y \mid 1)}.$$
$Z(W)$ 越大，channel 越難區分 inputs 0 與 1  
對 binary linear code 使用 maximum-likelihood decoding，並把平手算作 failure  
由 linearity 與 channel symmetry，只需要分析傳送 all-zero codeword 的情形

![alt text](image-5.png)

固定傳送的 codeword $c$ 與 competitor $c'$  
decoder 在兩者之間的選擇就是 binary hypothesis testing  
對任意 $a \ge 0$ 與 $b > 0$，有  
$$\mathbf 1\{a > b\} \le \sqrt{\frac{a}{b}}.$$
若平手也算 failure，將 $>$ 換成 $\ge$，不等式仍然成立

![alt text](image-6.png)

令 $a = W^n(y^n \mid c')$、$b = W^n(y^n \mid c)$，套用上述 inequality  
$$\begin{aligned}
P(c \to c')
&= \sum_{y^n} W^n(y^n \mid c)\,\mathbf 1\!\left\{W^n(y^n \mid c') \ge W^n(y^n \mid c)\right\} \\
&\le \sum_{y^n} \sqrt{W^n(y^n \mid c)W^n(y^n \mid c')} \\
&= Z(W)^{d_H(c, c')}.
\end{aligned}$$
最後一個等號使用了 memoryless 性質  
兩個 codewords 相同的位置貢獻 1，不同的位置各貢獻 $Z(W)$

![alt text](image-7.png)

![alt text](image-8.png)

![alt text](image-9.png)

傳送 all-zero codeword 時，weight 為 $i$ 的 nonzero codeword 與它的距離是 $i$  
對所有 competitors 使用 union bound，得到  
$$P_{\mathrm{fail}} \le \sum_{i=1}^n A_i Z(W)^i = W_{\mathcal C}(1, Z(W)) - 1.$$

對 $\operatorname{BSC}(p)$，假設 $0 \le p < 1/2$，則  
$$Z\!\left(\operatorname{BSC}(p)\right)
= \sqrt{(1 - p)p} + \sqrt{p(1 - p)} = 2\sqrt{p(1 - p)}.$$
代入同一個 black box，得到  
$$P_{\mathrm{fail}}^{\mathrm{BSC}}
\le W_{\mathcal C}\!\left(1, 2\sqrt{p(1 - p)}\right) - 1.$$
這個根號來自輸入 0 與 1 的兩個 conditional output distributions 之間的 overlap

Two-variable weight enumerator 對 BSC 的 error detection 還有一個精確用法  
令 $E$ 為 BSC 的 error vector；nonzero error 沒被偵測到，恰好表示 $E \in \mathcal C\setminus\{0\}$  
因此  
$$P_{\mathrm{undetected}}
= \sum_{c\in\mathcal C\setminus\{0\}}(1 - p)^{n-\operatorname{wt}(c)}p^{\operatorname{wt}(c)}
= W_{\mathcal C}(1 - p, p) - (1 - p)^n.$$
這是精確的 undetected-error probability，不是 maximum-likelihood decoding 的 block-error probability

## 試試看能不能算出各種 RM 的 Weight Enumerator

前面已經算出 $\mathrm{RM}(1, m)$ 的 weight enumerator  
接下來考慮 Hamming、extended Hamming 與 $\mathrm{RM}(2, m)$  
同時找出哪些 weights 會出現，以及每個 weight 有多少個 codewords

- 用 Greene theorem 連結矩陣的線性相依關係與 weight enumerator
- 由 duality 得到 MacWilliams identity，算出 Hamming 與 extended Hamming 的重量分布
- 用 quadratic form 的 canonical form，算出 $\mathrm{RM}(2, m)$ 的完整重量分布

在 $m = 4$ 時，$\mathrm{RM}(2, 4)$ 就是 extended Hamming code  
兩條計算路線會得到同一個多項式

### RM codes 的 duality

![alt text](image-10.png)

回顧 $\mathrm{RM}(r, m)$ 是次數至多 $r$ 的 multilinear polynomials 的取值向量  
對 $0 \le r < m$，有
$$\boxed{\mathrm{RM}(r, m)^\bot = \mathrm{RM}(m - r - 1, m).}$$
若 $\deg f \le r$、$\deg g \le m - r - 1$，則 $fg$ 化簡後次數小於 $m$  
每個次數小於 $m$ 的 monomial 都在偶數個點上取值為 1，所以
$$\sum_{x\in\mathbb F_2^m} f(x)g(x) = 0.$$
因此右邊包含在左邊；再比較 dimensions：
$$\sum_{j=0}^{m-r-1}\binom mj = 2^m - \sum_{j=0}^{r}\binom mj.$$
兩邊 dimension 相同，便得到等號

特別地，對 $m \ge 2$，
$$\mathrm{RM}(1, m)^\bot = \mathrm{RM}(m - 2, m) = \mathrm{EHam}_m,$$
其中 $\mathrm{EHam}_m$ 表示長度 $2^m$ 的 extended Hamming code  
其參數為 $[2^m, 2^m - m - 1, 4]$

另一方面，長度 $2^m - 1$ 的 Hamming code 記為 $\mathrm{Ham}_m$  
它的 dual 是 simplex code $\mathrm{Sim}_m$，參數為 $[2^m - 1, m, 2^{m-1}]$  
我們已知道這兩個 dual codes 的重量分布  
需要的是把 $W_{\mathcal C}$ 轉換成 $W_{\mathcal C^\bot}$ 的公式

## Matroids 與 Greene Theorem

### Generator matrix 的 column matroid

令 $G$ 是 $[n, k]_q$ code $\mathcal C$ 的 generator matrix，且 $\operatorname{rank}G = k$  
以座標集合 $E = \{1, \ldots, n\}$ 當作 ground set  
對 $A \subseteq E$，令 $G_A$ 是保留 $A$ 中各欄得到的矩陣，定義
$$\rho(A) = \operatorname{rank}G_A.$$
哪些欄的集合線性獨立，決定了 $G$ 的 **column matroid**，記為 $M$  
其 rank function 就是 $\rho$，而 $\rho(E) = k$

把 $G$ 左乘可逆矩陣，只是更換 code 的基底  
所有 $\rho(A)$ 都不變，所以這個 matroid 只依賴 code 與座標標記

### Tutte polynomial：同時記錄缺少的 rank 與相依關係

對每個 $A \subseteq E$，考慮兩個非負整數：

- $k - \rho(A)$：還缺多少個獨立方向，才能張成全部 $k$ 維空間
- $|A| - \rho(A)$：$A$ 中各欄之間，線性關係空間的 dimension

把所有子集合的資訊加總，得到 **Tutte polynomial**：
$$T_M(u, v) = \sum_{A\subseteq E}(u - 1)^{k-\rho(A)}(v - 1)^{|A|-\rho(A)}.$$
這是由 generator matrix 的欄相依結構定義的多項式  
接下來的 theorem 把它與 codewords 的 weight 連起來

### Greene theorem

對 $\mathbb F_q$ 上的 $[n, k]_q$ linear code $\mathcal C$，令 $M$ 為其 column matroid  
**Greene theorem** 給出
$$\boxed{W_{\mathcal C}(X, Y) = (X - Y)^k Y^{n-k}
T_M\!\left(\frac{X + (q - 1)Y}{X - Y}, \frac{X}{Y}\right).}$$
這裡沿用 $W_{\mathcal C}(X, Y) = \sum_{c\in\mathcal C}X^{n-\operatorname{wt}(c)}Y^{\operatorname{wt}(c)}$  
右邊化簡後是多項式；在 $X = Y$ 或 $Y = 0$ 時，用化簡後的式子代入

因此 codewords 的 weight distribution，可以由各座標子集合的 rank 資訊決定  
本節使用這個 theorem 的結論，接著看例子與 duality 的應用

### 例子：$[3, 2, 2]$ SPC code

取
$$G = \begin{pmatrix}1 & 0 & 1 \\ 0 & 1 & 1\end{pmatrix}.$$
三個欄向量都非零，任兩欄獨立，三欄的 rank 是 2  
依 $|A| = 0, 1, 2, 3$ 分組，得到
$$T_M(u, v) = (u - 1)^2 + 3(u - 1) + 3 + (v - 1) = u^2 + u + v.$$
代入 Greene theorem，得到
$$W_{\mathcal C}(X, Y) = Y(X + Y)^2 + Y(X - Y)(X + Y) + X(X - Y)^2
= X^3 + 3XY^2.$$
這正好對應 $000, 011, 101, 110$ 四個 codewords

### Dual matroid 與 dual code

Matroid $M$ 的 basis 是大小為 $k$ 的獨立欄集合  
**Dual matroid** $M^*$ 的 bases 定義為這些 bases 在 $E$ 中的補集合  
其 rank function 為
$$\rho^*(A) = |A| - k + \rho(E \setminus A).$$
對 linear code，$\mathcal C^\bot$ 的 column matroid 就是 $M^*$  
沿用本講義慣例，$G$ 是 $k \times n$、$H$ 是 $n \times (n - k)$  
因此這裡取 $H^\top$ 的 column matroid，且 $GH = 0$

這個對應可從 projection 看出來  
令 $\pi_A$ 表示只保留 $A$ 中的座標  
$\pi_A(\mathcal C^\bot)$ 在 $\mathbb F_q^A$ 中的正交補  
就是 $\mathcal C$ 中 support 包含在 $A$ 的 codewords，限制到 $A$ 的結果  
這些 codewords 是 $\pi_{E\setminus A}|_{\mathcal C}$ 的 kernel，其 dimension 是 $k - \rho(E \setminus A)$  
所以
$$\dim \pi_A(\mathcal C^\bot) = |A| - k + \rho(E \setminus A) = \rho^*(A).$$
把 rank 公式代入 Tutte polynomial，並將 $A$ 換成 $E \setminus A$，得到
$$\boxed{T_{M^*}(u, v) = T_M(v, u).}$$

### MacWilliams identity

對 $\mathcal C^\bot$ 使用 Greene theorem，再交換 Tutte 的兩個變數：
$$W_{\mathcal C^\bot}(X, Y) = (X - Y)^{n-k}Y^k
T_M\!\left(\frac{X}{Y}, \frac{X + (q - 1)Y}{X - Y}\right).$$
另一方面，令 $U = X + (q - 1)Y$、$V = X - Y$，則
$$U - V = qY,\qquad U + (q - 1)V = qX.$$
把 $U, V$ 代入 $\mathcal C$ 的 Greene 公式，得到
$$W_{\mathcal C}(U, V) = q^kY^k(X - Y)^{n-k}
T_M\!\left(\frac{X}{Y}, \frac{X + (q - 1)Y}{X - Y}\right).$$
比較兩式，就能消去相同的 Tutte polynomial

因此得到 **MacWilliams identity**：
$$\boxed{W_{\mathcal C^\bot}(X, Y) = \frac{1}{|\mathcal C|}
W_{\mathcal C}\bigl(X + (q - 1)Y, X - Y\bigr).}$$
對 binary code，$q = 2$，公式化成
$$W_{\mathcal C^\bot}(X, Y) = \frac{1}{2^k}W_{\mathcal C}(X + Y, X - Y).$$
已知任一邊的完整 weight enumerator，就可以算出另一邊  
當 $\mathcal C$ 的 dimension 很小而 $\mathcal C^\bot$ 很大時  
這個轉換能把大 code 的枚舉問題，交給小 code 的結構處理

## Hamming Codes 的完整重量分布

### 從 simplex code 算 Hamming code

令 $m \ge 2$、$n = 2^m - 1$  
Simplex code 由 $x \mapsto a \cdot x$ 在所有 $x \ne 0$ 的點上求值得到  
$a = 0$ 給出零向量，其餘 $2^m - 1$ 個 codewords 的 weight 都是 $2^{m-1}$  
所以
$$W_{\mathrm{Sim}_m}(X, Y) = X^n + (2^m - 1)X^{2^{m-1}-1}Y^{2^{m-1}}.$$
因為 $\mathrm{Ham}_m = \mathrm{Sim}_m^\bot$，由 MacWilliams identity 得到
$$\boxed{W_{\mathrm{Ham}_m}(z) = \frac{(1 + z)^n + (2^m - 1)(1 + z)^{2^{m-1}-1}(1 - z)^{2^{m-1}}}{2^m}.}$$
這裡先使用二變數公式，再代入 $X = 1$、$Y = z$

以 $m = 3$ 為例，simplex code 有 8 個 codewords  
其 weight enumerator 是 $1 + 7z^4$  
對偶的 Hamming $[7, 4, 3]$ code 得到
$$W_{\mathrm{Ham}_3}(z) = \frac{(1 + z)^7 + 7(1 + z)^3(1 - z)^4}{8}
= 1 + 7z^3 + 7z^4 + z^7.$$
所以 weight 為 3 與 4 的 codewords 各有 7 個  
代入 $z = 1$ 得到 $16 = 2^4$，最小正次方是 3  
也同時核對了 dimension 與 minimum distance

### 從一階 RM 算 extended Hamming code

令 $N = 2^m$，一階 RM 的二變數 weight enumerator 為
$$W_{\mathrm{RM}(1,m)}(X, Y) = X^N + (2^{m+1} - 2)X^{N/2}Y^{N/2} + Y^N.$$
因為其 dual 是 $\mathrm{EHam}_m$，所以
$$\boxed{W_{\mathrm{EHam}_m}(z) = \frac{(1 + z)^N + (2^{m+1} - 2)(1 - z^2)^{N/2} + (1 - z)^N}{2^{m+1}}.}$$
奇數次項全部消失，符合 extended Hamming code 的 overall parity check  
$m = 3$ 時得到 $1 + 14z^4 + z^8$  
$m = 4$ 時得到 $[16, 11, 4]$ code，稍後會從二次型再算一次

### 加上 overall parity 的另一種核對

Hamming codeword 的 weight 若為 $w$，加上 overall parity 後變成
$$w + (w \bmod 2).$$
令 $A_w$ 是 $\mathrm{Ham}_m$ 的 weight 分布，$\widetilde A_w$ 是 $\mathrm{EHam}_m$ 的分布  
則
$$\widetilde A_{2j} = A_{2j-1} + A_{2j},\qquad \widetilde A_{2j+1} = 0,$$
其中超出範圍的 $A_w$ 視為 0  
例如 $1 + 7z^3 + 7z^4 + z^7$ 加上 overall parity  
就變成 $1 + 14z^4 + z^8$

## Quadratic Forms 與二階 RM Codes

### 二次多項式的取值向量

以下令 $m \ge 2$，所有係數與輸入都在 $\mathbb F_2$ 中  
每個次數至多 2 的 Boolean polynomial 都有唯一的 multilinear 表示
$$f(x) = Q(x) + \ell(x) + b,\qquad
Q(x) = \sum_{1\le i<j\le m}a_{ij}x_ix_j,\qquad
\ell(x) = \sum_{i=1}^m c_ix_i.$$
在 $\mathbb F_2^m$ 的全部點上求值，得到 $\mathrm{RM}(2, m)$  
其參數為
$$\left[2^m,\ 1 + m + \binom m2,\ 2^{m-2}\right].$$
我們要利用 $Q$ 的結構，算出每種 weight 的 codeword 數量

### Weight 可以改寫成正負號加總

令 $\operatorname{wt}(f)$ 表示 $f$ 的取值向量的 weight，並定義
$$S(f) = \sum_{x\in\mathbb F_2^m}(-1)^{f(x)}.$$
$f(x) = 0$ 貢獻 $+1$，$f(x) = 1$ 貢獻 $-1$，因此
$$S(f) = 2^m - 2\operatorname{wt}(f),\qquad
\boxed{\operatorname{wt}(f) = \frac{2^m - S(f)}{2}.}$$
有限體中的值 $f(x) \in \{0, 1\}$ 在指數中視為整數  
這個加總與 weight 都是在整數中計算  
接下來將 $f$ 化成容易計算 $S(f)$ 的形式

### Polarization：從二次型得到雙線性形式

對二次部分 $Q$，定義 **polar form**
$$B_Q(x, y) = Q(x + y) + Q(x) + Q(y).$$
展開後得到
$$B_Q(x, y) = \sum_{i<j}a_{ij}(x_iy_j + x_jy_i) = xAy^\top,$$
其中 $A_{ij} = A_{ji} = a_{ij}$（$i < j$），且 $A_{ii} = 0$  
這裡 $x, y$ 沿用 row-vector 慣例  
$B_Q$ 對兩個輸入各自線性，且 $B_Q(x, x) = 0$  
滿足後一個條件的雙線性形式稱為 **alternating form（交替形式）**

例如 $Q(x) = x_1x_2 + x_2x_3$，則
$$A = \begin{pmatrix}0 & 1 & 0 \\ 1 & 0 & 1 \\ 0 & 1 & 0\end{pmatrix}.$$
這個矩陣的 rank 是 2  
取 $u_1 = x_1 + x_3$、$u_2 = x_2$、$u_3 = x_3$，就有
$$Q(x) = u_1u_2.$$
可逆線性換座標只是重新排列全部 $2^m$ 個輸入點  
所以 $S(f)$ 與 $\operatorname{wt}(f)$ 都不變  
這就是 canonical form 能用來計數的原因

### Alternating form 的 canonical form

若 $B \ne 0$，可選到 $e, h$ 使得 $B(e, h) = 1$  
在 $\operatorname{span}\{e, h\}$ 上，矩陣是
$$J = \begin{pmatrix}0 & 1 \\ 1 & 0\end{pmatrix}.$$
對任意向量 $v$，令
$$v' = v + B(v, h)e + B(v, e)h.$$
直接計算可得 $B(v', e) = B(v', h) = 0$  
因此可以把這個二維區塊與其正交補分開，再對正交補重複同樣步驟

重複分解後，存在可逆矩陣 $P$，使
$$PAP^\top = \operatorname{diag}(\underbrace{J, \ldots, J}_{s\text{ 個}}, 0_{m-2s}).$$
所以 alternating form 的 rank 必定是偶數，記為 $2s$  
其 **radical** 是
$$R = \{v : B(v, w) = 0\text{ 對所有 }w\},\qquad \dim R = m - 2s.$$
前 $2s$ 個方向成對配合，剩下的方向與所有向量都正交  
這個分解同時給出 canonical form 的建構方法與 rank 的意義

### 把 canonical form 帶回多項式

令 $x = uP$，在新座標下，$Q(uP)$ 的 polar form 與 $\sum_{i=1}^s u_{2i-1}u_{2i}$ 相同  
兩者之差的 polar form 為零，且在原點取值為零，所以差是一個線性函數  
將這個線性函數併入原有的 $\ell$，便得到
$$f(uP) = \sum_{i=1}^s u_{2i-1}u_{2i} + \sum_{j=1}^m d_ju_j + b.$$
這裡保留所有線性項與常數項，便能處理每一個 codeword

以下的 $2s$ 一律表示 polar form $B_Q$ 的 rank  
固定 $Q$ 後，當 $\ell$ 遍歷全部線性函數，新係數 $(d_1, \ldots, d_m)$ 也恰好遍歷 $\mathbb F_2^m$

### 每個二變數區塊只需要算一次

對 $\alpha, \beta \in \mathbb F_2$，完成乘積：
$$ab + \alpha a + \beta b = (a + \beta)(b + \alpha) + \alpha\beta.$$
平移 $a, b$ 不改變加總範圍，因此
$$\sum_{a,b\in\mathbb F_2}(-1)^{ab+\alpha a+\beta b}
= (-1)^{\alpha\beta}\sum_{a,b\in\mathbb F_2}(-1)^{ab}
= 2(-1)^{\alpha\beta}.$$
每一對變數都貢獻 $+2$ 或 $-2$  
而單一未配對變數的加總是
$$\sum_{a\in\mathbb F_2}(-1)^{da} = \begin{cases}2, & d = 0, \\ 0, & d = 1.\end{cases}$$

### 從區塊乘積得到所有可能的 weights

因為不同區塊使用不同變數，多重加總可拆成乘積  
若某個 $j > 2s$ 滿足 $d_j = 1$，則 $S(f) = 0$  
若所有 $d_j = 0$（$j > 2s$），則
$$S(f) = (-1)^{b+\sum_{i=1}^sd_{2i-1}d_{2i}}2^s2^{m-2s}
= (-1)^{b+\sum_{i=1}^sd_{2i-1}d_{2i}}2^{m-s}.$$
由 $\operatorname{wt}(f) = (2^m - S(f))/2$，得到可能的 weights：
$$\boxed{2^{m-1},\qquad 2^{m-1} - 2^{m-s-1},\qquad 2^{m-1} + 2^{m-s-1}.}$$
當 $2s = m$ 時，沒有未配對變數，因此不會出現 $S(f) = 0$ 的情形  
$s = 0$ 時，後兩個 weights 是 0 與 $2^m$，對應常數函數

### Canonical form 實際算了什麼？

取 $m = 4$，比較幾個函數：

| $f$ | Polar rank $2s$ | $S(f)$ | Weight |
|---|---:|---:|---:|
| $x_1x_2$ | 2 | $2 \cdot 2^2 = 8$ | 4 |
| $x_1x_2 + x_3$ | 2 | $2 \cdot 0 \cdot 2 = 0$ | 8 |
| $x_1x_2 + 1$ | 2 | $-8$ | 12 |
| $x_1x_2 + x_3x_4$ | 4 | $2 \cdot 2 = 4$ | 6 |
| $x_1x_2 + x_3x_4 + 1$ | 4 | $-4$ | 10 |

Polar rank 決定可能的偏離量，線性項與常數項決定實際落在哪個 weight

### 固定二次部分，計算各 weight 的數量

固定 $Q$，且 $\operatorname{rank}B_Q = 2s$  
讓 $\ell$ 與 $b$ 自由變動，共有 $2^{m+1}$ 個函數，形成 coset $Q + \mathrm{RM}(1, m)$  
未配對方向上的係數全為零時，前 $2s$ 個線性係數有 $2^{2s}$ 種選擇  
對每組選擇，$b = 0, 1$ 使 $S(f)$ 一正一負  
因此兩個偏離中央的 weights，各出現 $2^{2s}$ 次

其餘線性項在未配對方向上非零，weight 都是 $2^{m-1}$  
共有
$$2(2^m - 2^{2s}) = 2^{m+1} - 2^{2s+1}$$
個 codewords

令 $h = 2^{m-1}$、$\delta_s = 2^{m-s-1}$  
固定 polar rank 為 $2s$ 的二次部分，其 coset weight enumerator 是
$$\boxed{F_{m,s}(z) = 2^{2s}\bigl(z^{h-\delta_s} + z^{h+\delta_s}\bigr)
+ \bigl(2^{m+1} - 2^{2s+1}\bigr)z^h.}$$
檢查總數：$F_{m,s}(1) = 2^{m+1}$  
當 $s = 0$，二次部分是零，回到一階 RM：
$$F_{m,0}(z) = 1 + (2^{m+1} - 2)z^{2^{m-1}} + z^{2^m}.$$
剩下的工作，就是數出每個 polar rank 對應多少種 $Q$

## 計算各種 Polar Rank 的數量

### 二次部分與 alternating matrices 一一對應

固定原本的座標，二次部分只使用 $x_ix_j$（$i < j$）  
每組係數 $a_{ij}$ 恰好對應一個對稱且對角線為零的矩陣 $A$  
所以二次部分共有 $2^{\binom m2}$ 種  
令
$$N_{m,s} = |\{A \in \mathbb F_2^{m\times m} : A^\top = A,\ A_{ii} = 0,\ \operatorname{rank}A = 2s\}|.$$
要數 $N_{m,s}$，可以分成兩個選擇：

1. 選擇 dimension 為 $m - 2s$ 的 radical $R$
2. 在 $\mathbb F_2^m/R$ 上選擇 nondegenerate alternating form

### Gaussian binomial coefficient：數子空間

$\mathbb F_2^m$ 中 dimension 為 $d$ 的子空間數目是
$$\begin{bmatrix}m \\ d\end{bmatrix}_2
= \prod_{i=0}^{d-1}\frac{2^m - 2^i}{2^d - 2^i}.$$
分子是從 $\mathbb F_2^m$ 依序選 $d$ 個獨立向量的方法數  
分母是同一個 $d$ 維子空間內，有序基底的方法數  
每個子空間被重複數到相同次數，相除就得到答案

因此 radical 的選擇數是
$$\begin{bmatrix}m \\ m - 2s\end{bmatrix}_2 = \begin{bmatrix}m \\ 2s\end{bmatrix}_2.$$
$d = 0$ 時空乘積為 1，表示零子空間只有一個

### 數 nondegenerate alternating forms

在固定的 $2s$ 維空間上，令 $a_s$ 為 nondegenerate alternating forms 的數目  
每個有序基底 $(e_1, h_1, \ldots, e_s, h_s)$，都唯一指定一個在此基底下矩陣為 $\operatorname{diag}(J, \ldots, J)$ 的 form  
有序基底共有
$$|\mathrm{GL}(2s, 2)| = \prod_{i=0}^{2s-1}(2^{2s} - 2^i)$$
種；但同一個 form 會被許多基底重複表示  
對固定 form，滿足 $B(e_i, h_j) = \delta_{ij}$、$B(e_i, e_j) = B(h_i, h_j) = 0$ 的基底  
稱為 **symplectic basis**  
因此需要除以每個 form 的 symplectic bases 數目

令 $D_s$ 為固定 nondegenerate alternating form 的 symplectic bases 數目  
第一個向量 $e_1 \ne 0$ 有 $2^{2s} - 1$ 種選擇  
條件 $B(e_1, h_1) = 1$ 給出 $2^{2s-1}$ 種 $h_1$  
選好這一對後，其正交補是 $2s - 2$ 維，且限制後仍 nondegenerate  
因此
$$D_s = (2^{2s} - 1)2^{2s-1}D_{s-1},\qquad D_0 = 1.$$
展開得到
$$D_s = 2^{s^2}\prod_{i=1}^s(2^{2i} - 1).$$
每個 nondegenerate alternating form 都有相同數量的 symplectic bases

所以固定 $2s$ 維空間上的 form 數目是
$$a_s = \frac{|\mathrm{GL}(2s, 2)|}{D_s}
= 2^{s(s-1)}\prod_{i=1}^s(2^{2i-1} - 1).$$
給定 radical $R$ 後，每個 quotient 上的 nondegenerate form  
都唯一拉回成原空間上 radical 恰為 $R$ 的 alternating form  
因此
$$\boxed{N_{m,s} = \begin{bmatrix}m \\ 2s\end{bmatrix}_2
2^{s(s-1)}\prod_{i=1}^s(2^{2i-1} - 1),\qquad 0 \le s \le \lfloor m/2\rfloor.}$$
$s = 0$ 時空乘積為 1，因此 $N_{m,0} = 1$  
所有 ranks 加起來應滿足 $\sum_s N_{m,s} = 2^{\binom m2}$

### 二階 RM 的完整 weight enumerator

每個 $f = Q + \ell + b$ 的 multilinear 表示唯一  
不同的二次部分 $Q$ 給出互不重疊的 cosets  
將相同 polar rank 的 cosets 分組，得到
$$\boxed{W_{\mathrm{RM}(2,m)}(z)
= \sum_{s=0}^{\lfloor m/2\rfloor}N_{m,s}F_{m,s}(z).}$$
其中 $F_{m,s}$ 記錄一個 coset 內的 weights，$N_{m,s}$ 記錄這類 cosets 的數量  
代入 $z = 1$，總 codeword 數是
$$\sum_s N_{m,s}2^{m+1} = 2^{\binom m2}2^{m+1}
= 2^{1+m+\binom m2}.$$
這與 $\mathrm{RM}(2, m)$ 的 dimension 完全一致

### 例子：$\mathrm{RM}(2, 3)$ 是 SPC code

$m = 3$ 時，polar rank 只有 0 與 2  
其數量為 $N_{3,0} = 1$、$N_{3,1} = 7$，所以
$$F_{3,0}(z) = 1 + 14z^4 + z^8,\qquad F_{3,1}(z) = 4z^2 + 8z^4 + 4z^6.$$
相加得到
$$W_{\mathrm{RM}(2,3)}(z) = 1 + 28z^2 + 70z^4 + 28z^6 + z^8.$$
這正是長度 8 的所有偶數 weight 向量：
$$W_{\mathrm{RM}(2,3)}(z) = \frac{(1 + z)^8 + (1 - z)^8}{2}.$$
因此二次型的計數與 SPC code 的描述相符

### 例子：$\mathrm{RM}(2, 4)$ 是 extended Hamming code

$m = 4$ 時，polar rank 可能是 0、2、4  
各類二次部分的數量為
$$N_{4,0} = 1,\qquad N_{4,1} = 35,\qquad N_{4,2} = 28.$$
總數 $1 + 35 + 28 = 64 = 2^{\binom42}$  
對應的 coset weight enumerators 是
$$\begin{aligned}
F_{4,0}(z) & = 1 + 30z^8 + z^{16}, \\
F_{4,1}(z) & = 4z^4 + 24z^8 + 4z^{12}, \\
F_{4,2}(z) & = 16z^6 + 16z^{10}.
\end{aligned}$$

因此
$$\begin{aligned}
W_{\mathrm{RM}(2,4)}(z)
& = F_{4,0}(z) + 35F_{4,1}(z) + 28F_{4,2}(z) \\
& = 1 + 140z^4 + 448z^6 + 870z^8 + 448z^{10} + 140z^{12} + z^{16}.
\end{aligned}$$
係數總和為 $2{,}048 = 2^{11}$，最小非零 weight 是 4  
這與 $[16, 11, 4]$ 的參數一致  
同時，$f \mapsto f + 1$ 使 weight $w$ 變成 $16 - w$  
所以係數以中央 weight 8 為中心對稱

另一方面，$\mathrm{RM}(2, 4) = \mathrm{EHam}_4 = \mathrm{RM}(1, 4)^\bot$  
Greene theorem 與 duality 給出的 MacWilliams 公式是
$$W_{\mathrm{EHam}_4}(z) = \frac{(1 + z)^{16} + 30(1 - z^2)^8 + (1 - z)^{16}}{32}.$$
展開後，恰好得到同一組係數

在這個例子中，canonical form 把 quadratic polynomials 分成可計數的區塊  
Greene theorem 則把 column dependence 與 weight enumerator 連起來  
兩種結構都能讓我們在不逐一列出 codewords 的情況下，得到完整重量分布

## 週五作業

### RM codes 上的三種變換

令 $X_1, \ldots, X_m$ 為形式變數，並記 $X = (X_1, \ldots, X_m)$  
取 $f(X) \in \mathbb F_2[X_1, \ldots, X_m]$，其中 $\deg f \le r$  
取值點則記為 $a = (a_1, \ldots, a_m) \in \mathbb F_2^m$  
$f$ 對應的 codeword $c \in \mathcal C = \mathrm{RM}(r, m)$ 定義為  
$$c_a = f(a_1, \ldots, a_m).$$
給定取值點集合上的 bijection $T : \mathbb F_2^m \to \mathbb F_2^m$，定義  
$$(T \cdot c)_a = c_{T(a)}.$$
因為 $T$ 排列取值點，所以 $T \cdot c$ 只是重新排列 $c$ 的座標  
下面會對每種 $T$ 分別用 substitution 定義 polynomial $T * f$  
1. 每次定義後，先驗證 $(T \cdot c)_a = (T * f)(a)$

#### Translation：$a \mapsto a + b$

固定 $b \in \mathbb F_2^m$，令 $T_b(a) = a + b$  
此時對應的 substitution 是  
$$(T_b * f)(X) = f(X_1 + b_1, \ldots, X_m + b_m).$$
2. 證明 $T_b$ 是取值點集合上的 bijection，並寫出它的 inverse  
3. 先取 degree 為 $s$ 的 multilinear monomial  
$$X_{j_1}\cdots X_{j_s}.$$
做 substitution $X_i \mapsto X_i + b_i$ 後展開；  
注意到每一項的 degree 都不超過 $s$  
4. 再對一般的 $f$ 使用上一題，證明 $\deg(T_b * f) \le r$  
5. 得到結論：$c \in \mathcal C$ 時，$T_b \cdot c \in \mathcal C$

#### Swap：$(a_i, a_j) \mapsto (a_j, a_i)$

定義 $(S_{ij} * f)(X)$ 為把 $f$ 中的形式變數 $X_i$ 與 $X_j$ 交換後得到的 polynomial  
6. 說明 $S_{ij}$ 如何作用在取值點集合上，證明它是 bijection
7. 寫出 inverse  
8. 證明 $\deg(S_{ij} * f) \le r$，並推出 $S_{ij} \cdot c \in \mathcal C$

#### Shear：$(a_i, a_j) \mapsto (a_i, a_j + a_i)$

定義 $(E_{ij} * f)(X)$ 為對 $f$ 做 substitution $X_j \mapsto X_j + X_i$  
9. 說明 $E_{ij}$ 如何作用在取值點集合上，找出 $E_{ij}$ 的 inverse  
10. 分別考慮不含 $X_j$、含 $X_j$ 但不含 $X_i$，以及同時含的 monomials，
證明 degree 不會增加？  
11. 證明 $\deg(E_{ij} * f) \le r$，所以 $E_{ij} \cdot c \in \mathcal C$

### 週一作業

12. 綜合以上結果，證明 $\mathrm{RM}(r, m)$ 對 translation、swap 與 shear 封閉  
13. 這三種變換的 inverse 也屬於同一類  
說明為什麼它們不只把 $\mathcal C$ 映入自身，而是把 $\mathcal C$ 映到自身  
因此三種變換都屬於 code 的 automorphism group $\operatorname{Aut}(\mathcal C)$

#### 從 generators 到 $\operatorname{AGL}(m, 2)$

14. 在 $\mathbb F_2$ 上對 invertible matrix 做 Gaussian elimination 的時候，
除了交換兩個 coordinates 與把一個 coordinate 加到另一個 coordinate，是否還需要 scaling？  
據此證明 swap 與 shear 生成 $\operatorname{GL}(m, 2)$  
15. 證明每個 affine bijection 都可以寫成  
$$a \longmapsto Aa + b,\qquad A \in \operatorname{GL}(m, 2),\quad b \in \mathbb F_2^m.$$
並說明它可以由前面的 translation、swap 與 shear 組成  
16. 最後得到   
$$\boxed{\operatorname{AGL}(m, 2) \le \operatorname{Aut}(\mathrm{RM}(r, m)).}$$
